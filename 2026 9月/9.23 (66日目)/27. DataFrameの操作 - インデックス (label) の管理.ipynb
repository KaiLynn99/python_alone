```python
# インデックス操作（ラベルインデックスであり、位置番号ではない）

# 現在のインデックスを確認
import pandas as pd


df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル"],
    "点数": [85, 90, 75]
})


print(df)

print(df.index)
```

<br>

```python
# インデックスの値をリストとして返す
print(df.index.tolist())

# インデックス名を確認、設定されていない場合はNone
print(df.index.name)
```

<br>

```python
# インデックス名を設定

df.index.name = "学生区分"

df
```

<br>

```python
# インデックスを変更

df.index = ["学生1", "学生2", "学生3"]

df
```

<br>

#### - `set_index`：列 → インデックス

```python
# DataFrameを確認

df = pd.DataFrame({
    "学生ID": [101, 102, 103],
    "名前": ["千尋", "キキ", "ハウル"],
    "点数": [85, 90, 75]
})


df
```

<br>

```python
# 列をインデックスに変更

df2 = df.set_index("学生ID")

df2
```

<br>

```python
# inplace=True → 元のDataFrame自体を直接変更

df.set_index("学生ID", inplace=True)  # 「学生ID」がインデックス名になる

df
```

<br>

#### - `reset_index`：インデックス → 列

```python
# DataFrameを確認

df = pd.DataFrame({
    "学生ID": [101, 102, 103],
    "名前": ["千尋", "キキ", "ハウル"],
    "点数": [85, 90, 75]
})

df
```

<br>

```python
# インデックスを列に変更

# namesを指定しない場合、列名はデフォルトで「index」になる

df2 = df.reset_index(names="番号")

df2
```

<br>

```python
# 既存のインデックスを削除

df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル"]
}, index=[10, 20, 30])

df
```

<br>

```python
# reset_index()のデフォルトでは既存のインデックスを列に移動するが、
# drop=Trueを指定すると既存のインデックスを削除する

df2 = df.reset_index(drop=True)

df2
```

<br>

```python
# インデックスを並べ替え

df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル"]
}, index=[30, 10, 20])

df
```

<br>

```python
# インデックスを昇順に並べ替え

df2 = df.sort_index()

df2
```
