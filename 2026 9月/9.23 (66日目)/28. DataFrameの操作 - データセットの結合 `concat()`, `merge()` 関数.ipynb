 共通キーを基準に（SQLのJOINに類似）データを結合する場合は、merge()またはjoin()を使用する。

- `join="outer"` → すべての列（またはインデックス）を維持する  
- `join="inner"` → 共通する列（またはインデックス）のみ維持する  
- SQLの`JOIN`のように共通キーを基準に結合する機能ではない。

<br>

```python
import pandas as pd


df1 = pd.DataFrame({
    "名前": ["千尋", "キキ"],
    "点数": [80, 90]
})


df2 = pd.DataFrame({
    "名前": ["ハウル", "ソフィー"],
    "点数": [85, 95]
})


display(df1)
display(df2)
```

<br>

#### - `ignore_index=False`：デフォルト値 <br> → インデックスを`維持`

```python
# axis=0 → 上下方向に結合（基本）→ 行を追加

df = pd.concat(
    [df1, df2],
    ignore_index=False,
    axis=0
) 

display(df)
```

<br> 

#### - `ignore_index=True` <br> → インデックスを`リセット`

```python
df = pd.concat(
    [df1, df2],
    ignore_index=True,
    axis=0
)

display(df)
```

<br>

#### - `axis=1` <br> → 省略した場合の`デフォルト値`は `0`

```python
# axis=1 → 左右に結合 → 列を追加

df = pd.concat([df1, df2], axis=1)

display(df)
```

<br>

#### ※ `join="outer"` vs `join="inner"`

`join`は、**全体を維持するか、共通部分だけを維持するか**を決定する。

```python
df1 = pd.DataFrame({
    "名前": ["千尋", "キキ"],
    "点数": [80, 90]
})


df2 = pd.DataFrame({
    "名前": ["ハウル", "ソフィー"],
    "等級": ["A", "B"]
})


display(df1)
display(df2)
```

#### → 2つのDataFrameで共通する列は、`「名前」`列のみである。

<br>

#### - join="outer"の例 <br> (両方のDataFrameに存在するすべての列を維持する)

```python
# joinを省略した場合、デフォルト値は"outer"

df = pd.concat(
    [df1, df2],
    join="outer",
    ignore_index=True
)

display(df)
```

<br>

#### - join="inner"の例 <br> (両方のDataFrameに共通して存在する列のみ維持する)

```python
df = pd.concat(
    [df1, df2],
    join="inner",
    ignore_index=True
)

display(df)
```

<br>

#### - `axis=1`：列方向に結合

```python
import pandas as pd


df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー"],
    "点数": [90, 80, 95, 70]
})


grade = pd.DataFrame({
    "等級": ["A", "B", "A", "C"]
})


# axis=1 → インデックスを基準に左右へ結合
result = pd.concat([df, grade], axis=1)

display(result)
```

<br>

#### - 行数（インデックス）が一致しない場合 <br> → 存在しない値がNaNで補われる

```python
df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー"],
    "点数": [90, 80, 95, 70]
})


grade2 = pd.DataFrame({
    "等級": ["A", "B", "A"]
})


result = pd.concat([df, grade2], axis=1)

display(result)
```

<br>

```python
# `join="inner"`：両方に共通するインデックスだけを残して結合する。

pd.concat([df, grade2], axis=1, join="inner")
```

#### # merge()によるDataFrameの結合

2つのDataFrameを`顧客ID`をキーとして結合する。

結合方法は`how`引数で指定し、`inner`・`left`・`right`・`outer`の4種類を比較する。

```python
import pandas as pd


df_customer = pd.DataFrame({
    "顧客ID": [101, 102, 103, 104],
    "名前": ["千尋", "ハウル", "ソフィー", "アシタカ"]
})

df_score = pd.DataFrame({
    "顧客ID": [101, 102, 103, 105],
    "点数": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)
```

<br>

#### 1. inner結合

`inner`は、両方のDataFrameに共通して存在するキーのデータだけを結合する。

この例では、両方に存在する`顧客ID`の`101・102・103`だけが結果に残る。

```python
result = pd.merge(
    df_customer,
    df_score,
    on="顧客ID",
    how="inner"  # inner（デフォルト）、left、right、outer
)

display(result)
```

<br>

#### 2. left結合

`left`は、左側のDataFrameを基準にすべてのデータを残す。

右側に一致するデータがない場合は、該当する値が`NaN`になる。

```python
result = pd.merge(
    df_customer,
    df_score,
    on="顧客ID",
    how="left"
)

display(result)
```

<br>

#### 3. right結合

`right`は、右側のDataFrameを基準にすべてのデータを残す。

左側に一致するデータがない場合は、該当する値が`NaN`になる。

```python
result = pd.merge(
    df_customer,
    df_score,
    on="顧客ID",
    how="right"
)

display(result)
```

<br>

#### 4. outer結合

`outer`は、左右両方のDataFrameに存在するすべてのキーを残す。

どちらか一方にしか存在しないデータは、対応する値が`NaN`になる。

```python
result = pd.merge(
    df_customer,
    df_score,
    on="顧客ID",
    how="outer"
)

display(result)
```

<br>

#### ※ 結合方法の違い

- `inner`：両方に存在するデータのみ
- `left`：左側のデータをすべて残す
- `right`：右側のデータをすべて残す
- `outer`：左右両方のデータをすべて残す

<br>

※ pandasライブラリには、`merge()`関数と似た機能を持つ`DataFrame.merge()`、`DataFrame.join()`などのメソッドもある。

似た機能を持つ`merge()`メソッドや`join()`メソッドが存在するのは、使用する場面やコードの書き方が少し異なるためである。

基本的には、まず`merge()`関数を使えるようになれば十分である。
