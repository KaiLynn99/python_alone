 共通キーを基準に（SQLのJOINに類似）データを結合する場合は、merge()またはjoin()を使用する。

<br clear="left">

- `join="outer"` → すべての列（またはインデックス）を維持する  
- `join="inner"` → 共通する列（またはインデックス）のみ維持する  
- SQLの`JOIN`のように共通キーを基準に結合する機能ではない。

<br>

```python
import pandas as pd


df1 = pd.DataFrame({
    "名前": ["千尋", "キキ"],
    "点数": [80, 90]
})


df2 = pd.DataFrame({
    "名前": ["ハウル", "ソフィー"],
    "点数": [85, 95]
})


display(df1)
display(df2)
```

<br>

#### - `ignore_index=False`：デフォルト値 <br> → インデックスを`維持`

```python
# axis=0 → 上下方向に結合（基本）→ 行を追加

df = pd.concat(
    [df1, df2],
    ignore_index=False,
    axis=0
) 

display(df)
```

<br> 

#### - `ignore_index=True` <br> → インデックスを`リセット`

```python
df = pd.concat(
    [df1, df2],
    ignore_index=True,
    axis=0
)

display(df)
```

<br>

#### - `axis=1` <br> → 省略した場合の`デフォルト値`は `0`

```python
# axis=1 → 左右に結合 → 列を追加

df = pd.concat([df1, df2], axis=1)

display(df)
```

<br>

#### `join="outer"` vs `join="inner"`

`join`は、**全体を維持するか、共通部分だけを維持するか**を決定する。

```python
df1 = pd.DataFrame({
    "名前": ["千尋", "キキ"],
    "点数": [80, 90]
})


df2 = pd.DataFrame({
    "名前": ["ハウル", "ソフィー"],
    "等級": ["A", "B"]
})


display(df1)
display(df2)
```

#### → 2つのDataFrameで共通する列は、`「名前」`列のみである。

##### -  join="outer"の例 <br> (両方のDataFrameに存在するすべての列を維持する)

```python
# joinを省略した場合、デフォルト値は"outer"

df = pd.concat(
    [df1, df2],
    join="outer",
    ignore_index=True
)

display(df)
```

<br>

##### -  join="inner"の例 <br> (両方のDataFrameに共通して存在する列のみ維持する)

```python
df = pd.concat(
    [df1, df2],
    join="inner",
    ignore_index=True
)

display(df)
```

##### - `axis=1`：列方向に結合

```python
import pandas as pd


df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー"],
    "点数": [90, 80, 95, 70]
})


grade = pd.DataFrame({
    "等級": ["A", "B", "A", "C"]
})


# axis=1 → インデックスを基準に左右へ結合
result = pd.concat([df, grade], axis=1)

display(result)
```

<br>

##### 行数（インデックス）が一致しない場合 <br> → 存在しない値がNaNで補われる

```python
df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー"],
    "点数": [90, 80, 95, 70]
})


grade2 = pd.DataFrame({
    "等級": ["A", "B", "A"]
})


result = pd.concat([df, grade2], axis=1)

display(result)
```
