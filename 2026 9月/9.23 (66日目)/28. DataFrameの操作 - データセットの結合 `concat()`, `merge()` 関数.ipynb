## 1. `concat()` 関数

 : 同じ構造のデータを結合する

 - SQLのJOINとは異なる。
 - `concat`は単純に`軸を基準につなげるもの`であり、<br>
 共通キーを基準に（SQLのJOINに類似）データを結合する場合は、merge()またはjoin()を使用する。

<br clear="left">

- `join="outer"` → すべての列（またはインデックス）を維持する  
- `join="inner"` → 共通する列（またはインデックス）のみ維持する  
- SQLの`JOIN`のように共通キーを基準に結合する機能ではない。

<br>

```python
import pandas as pd


df1 = pd.DataFrame({
    "名前": ["千尋", "キキ"],
    "点数": [80, 90]
})


df2 = pd.DataFrame({
    "名前": ["ハウル", "ソフィー"],
    "点数": [85, 95]
})


display(df1)
display(df2)
```

<br>

#### - `ignore_index=False`：デフォルト値 <br> → インデックスを`維持`

```python
# axis=0 → 上下方向に結合（基本）→ 行を追加

df = pd.concat(
    [df1, df2],
    ignore_index=False,
    axis=0
) 

display(df)
```

<br> 

#### - `ignore_index=True` <br> → インデックスを`リセット`

```python
df = pd.concat(
    [df1, df2],
    ignore_index=True,
    axis=0
)

display(df)
```
