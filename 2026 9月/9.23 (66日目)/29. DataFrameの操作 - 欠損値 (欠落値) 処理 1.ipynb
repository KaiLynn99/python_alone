#### 欠損値の確認 - isna() / isnull()

`isna()`または`isnull()`を使用して、DataFrameの欠損値を確認できる。

- `None`：Pythonで「値が存在しない」ことを表すオブジェクト
- `NaN`：NumPy・pandasで主に数値データの欠損値を表す値

```python
import pandas as pd

data = {
    "名前": ["千尋", "ハウル", "ソフィー", "キキ", "アシタカ"],
    "点数": [85, None, 90, 75, None],
    "出席": [True, True, None, True, False]
}

df = pd.DataFrame(data)

df  # 数値列のNoneはNaNとして処理される場合がある
```

<br>

```python
print(df.isna())

print()

print(df.isnull())
```

<br>

<span style="color:#0070C0; font-weight:bold;">
※ Boolean値にsum()演算を行うと、True → 1、False → 0として計算される。<br>
→ sum()メソッドを利用すると、欠損値の個数を確認できる。
</span>
