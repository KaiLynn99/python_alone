#### ※ データ全体の構成

`Species` → 魚の種類を表す変数。  
分類問題では<strong>正解ラベル（Label）</strong>として使用する。

<br>

データには次の魚種が含まれている。

`Bream、Roach、Whitefish、Parkki、Perch、Pike、Smelt`

<br>

#### # Bream・Smeltを用いたKNNモデルの学習とテストデータによる予測・評価

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# データの読み込み
fish = pd.read_csv(
    "https://raw.githubusercontent.com/rickiepark/hg-mldl/master/fish.csv"
)

# BreamとSmeltのみを選択
fish = fish[fish["Species"].isin(["Bream", "Smelt"])]

# 入力データ（X）
X = fish[["Length", "Weight"]]

# 正解データ（y）
# Bream = 1、Smelt = 0
y = fish["Species"].map({
    "Bream": 1,
    "Smelt": 0
})

# KNNモデルの作成・学習
kn = KNeighborsClassifier(n_neighbors=5)
kn.fit(X, y);
```

<br>

テスト用kn.score(X_test, y_test) データセット

```python
# テストデータ
X_test = [
    # Bream 30個
    [32.0, 340],
    [33.0, 360],
    [34.0, 390],
    [35.0, 420],
    [36.0, 450],
    [37.0, 480],
    [38.0, 510],
    [39.0, 540],
    [40.0, 570],
    [41.0, 600],
    [42.0, 630],
    [43.0, 660],
    [44.0, 690],
    [45.0, 720],
    [46.0, 750],
    [47.0, 780],
    [48.0, 810],
    [49.0, 840],
    [50.0, 870],
    [51.0, 900],
    [52.0, 920],
    [53.0, 940],
    [54.0, 960],
    [55.0, 980],
    [56.0, 1000],
    [57.0, 1020],
    [58.0, 1040],
    [59.0, 1060],
    [60.0, 1080],
    [61.0, 1100],

    # Smelt 15個
    [9.0, 6],
    [10.0, 7],
    [11.0, 8],
    [12.0, 9],
    [13.0, 10],
    [14.0, 11],
    [15.0, 12],
    [16.0, 13],
    [17.0, 14],
    [18.0, 15],
    [19.0, 16],
    [20.0, 17],
    [21.0, 18],
    [22.0, 19],
    [23.0, 20]
]

# 実際の正解
y_test = [1] * 30 + [0] * 15
```

<br>

#### ⓪ 予測前の準備

`accuracy_score`を読み込み、X_testに`Length`・`Weight`の列名を付けてDataFrameに変換する。

```python
from sklearn.metrics import accuracy_score

X_test = pd.DataFrame(
    X_test,
    columns=["Length", "Weight"]
)
```

<br>

#### ① 新しいデータを予測

```python
prediction = kn.predict(X_test)

print(prediction)
print(accuracy_score(y_test, prediction))
```

<br>

#### ② `score()`で直接正解率を確認

```python
print(kn.score(X_test, y_test))
```
