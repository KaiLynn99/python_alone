## 1. ヒストグラム：1つのデータの分布と頻度を探索

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. データを準備
# ============================================================

# seabornが提供するtipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データを確認
print(tips.head())


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置するための大きな描画領域（Figure）を作成する
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域（Axes）を作成する
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. ヒストグラムを描画
# ============================================================

# total_bill列のデータを使用してヒストグラムを描画する
#
# hist()
# → 数値データを複数の区間（bin）に分け、
#   各区間にデータがいくつ存在するかを棒で表現する
#
# bins=10
# → データ全体を10個の区間に分ける
axes1.hist(tips["total_bill"], bins=10)


# ============================================================
# 5. グラフを設定
# ============================================================

# グラフのタイトル
axes1.set_title("Total Bill Distribution")

# x軸のラベル
axes1.set_xlabel("Total Bill")

# y軸のラベル
axes1.set_ylabel("Frequency")


# ============================================================
# 6. グラフの余白を調整
# ============================================================

# タイトルや軸ラベルなどが重ならないように
# グラフの余白を自動的に調整する
fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

plt.show()
```

## 2. 散布図

#### - `tip`, `total_bill` 変数

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. データを準備
# ============================================================

# seabornが提供するtipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データの先頭部分を確認
print(tips.head())


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置するための大きな描画領域（Figure）を作成する
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域（Axes）を作成する
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 散布図を描画
# ============================================================

# x軸 → 食事金額（total_bill）
# y軸 → チップ（tip）
#
# 1つの点が1人の顧客のデータを表す
#
# 散布図は2つの数値型変数の間にある
# 関係やパターンを確認するときに使用する
axes1.scatter(
    tips["total_bill"],
    tips["tip"]
)


# ============================================================
# 5. グラフを設定
# ============================================================

# グラフのタイトル
axes1.set_title("Total Bill vs Tip")

# x軸のラベル
axes1.set_xlabel("Total Bill")

# y軸のラベル
axes1.set_ylabel("Tip")


# ============================================================
# 6. グラフの余白を調整
# ============================================================

# タイトルや軸ラベルなどが見やすくなるように
# グラフの余白を自動的に調整する
fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

plt.show()
```

#### - `tip`, `total`, `bill`, `sex` 変数

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. データを準備
# ============================================================

# seabornが提供するtipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データを確認
print(tips.head())


# ============================================================
# 2. 性別ごとにデータを分割
# ============================================================

# 性別がMaleのデータのみを抽出する
male = tips.loc[tips["sex"] == "Male"]

# 性別がFemaleのデータのみを抽出する
female = tips.loc[tips["sex"] == "Female"]


# ============================================================
# 3. Figure全体を作成
# ============================================================

# グラフ全体を配置するための大きな描画領域（Figure）を作成する
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 4. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域を作成する
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 5. 散布図を描画
# ============================================================

# 男性データ
#
# x軸 → total_bill : 食事金額
# y軸 → tip        : チップ
#
# labelを指定すると凡例に表示できる
axes1.scatter(
    male["total_bill"],
    male["tip"],
    label="Male"
)


# 女性データ
axes1.scatter(
    female["total_bill"],
    female["tip"],
    label="Female"
)


# ============================================================
# 6. グラフを設定
# ============================================================

# グラフのタイトル
axes1.set_title("Total Bill vs Tip by Sex")

# x軸のラベル
axes1.set_xlabel("Total Bill")

# y軸のラベル
axes1.set_ylabel("Tip")

# 凡例を表示
axes1.legend()


# ============================================================
# 7. Figureを表示
# ============================================================

fig.tight_layout()

plt.show()
```

## 3. 箱ひげ図

#### - `tip` 変数

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. データを準備
# ============================================================

# seabornが提供するtipsデータセットを読み込む
tips = sns.load_dataset("tips")


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置するためのFigureを作成する
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域を作成する
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. tipの箱ひげ図を描画
# ============================================================

# tip列のデータ分布を箱ひげ図で表現する
axes1.boxplot(tips["tip"])


# ============================================================
# 5. グラフを設定
# ============================================================

# グラフのタイトル
axes1.set_title("Tip Distribution")

# y軸のラベル
axes1.set_ylabel("Tip")


# ============================================================
# 6. Figureを表示
# ============================================================

fig.tight_layout()

plt.show()
```

#### - `性別`、`tip` 変数

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. データを準備
# ============================================================

# seabornが提供するtipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データを確認
print(tips.head())


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置するための大きな描画領域（Figure）を作成する
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域（Axes）を作成する
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 性別ごとのtipの箱ひげ図を描画
# ============================================================

# 男性のtipデータ
male_tip = tips.loc[tips["sex"] == "Male", "tip"]

# 女性のtipデータ
female_tip = tips.loc[tips["sex"] == "Female", "tip"]


# 男性と女性のtipの分布を箱ひげ図で比較する
axes1.boxplot(
    [male_tip, female_tip],
    
    # 現在のMatplotlibではlabelsの代わりに
    # tick_labelsを使用する
    tick_labels=["Male", "Female"]
)


# ============================================================
# 5. グラフを設定
# ============================================================

# グラフのタイトル
axes1.set_title("Tip Distribution by Sex")

# x軸のラベル
axes1.set_xlabel("Sex")

# y軸のラベル
axes1.set_ylabel("Tip")


# ============================================================
# 6. グラフの余白を調整
# ============================================================

fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

plt.show()
```
