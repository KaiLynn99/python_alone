## 1. ネットワーク分析

## 1) 伝統的統計 vs ネットワーク分析

#### **「何を分析の中心に置くのか？」**

| 区分 | **伝統的統計** | **ネットワーク分析** |
|---|---|---|
| **分析対象** | `個々の観測値` | `個体と個体の間の関係` |
| **関心対象** | `変数` | `ノード（Node）とエッジ（Edge）` |
| **基本質問** | 「Aの値はいくらか？」 | 「Aは誰とつながっているのか？」 |
| **データ構造** | `行 × 列` | `接続構造` |
| **主要な関係** | `変数間の相関関係` | `個体間の接続関係` |
| **代表的な分析** | `平均、分散、回帰、相関、検定` | `中心性、コミュニティ、連結性、経路` |
| **代表的な指標** | 平均、標準偏差、R²、p-value | Degree、Betweenness、Closeness、PageRank |
| **主な可視化** | `棒グラフ、ヒストグラム、散布図` | `ネットワークグラフ` |
| **代表的なツール** | R、Python、SPSS、Excel | `Gephi、NetworkX、UCINET` |
| **中心となる観点** | `変数中心` | `関係中心` |

- データの中に存在する「関係」を見つけて分析する方法。
- 新聞記事は単なる文字の集まりではなく、人物、企業、地域、事件、キーワードなどが互いにつながっている。

## 2. ネットワークの2つの核心要素

#### 1) `ノード（Node）＝ 対象`

新聞記事の場合、

- 人物
- 企業
- 地域
- 単語
- 事件

などが**ノード**になる。

#### 2) `エッジ（Edge）＝ 関係`

例えば、2つの企業が同じ記事に登場した場合、

## 3. 何が分かるのか？

ネットワークを描くことで、単純な記事内容だけでは見えなかったパターンを見つけることができる。

- **最も多くつながっているのは誰か？** → `中心性`
- **どの単語・人物が互いに強くつながっているのか？** → `共起回数` <br>
(※ 本ノートでは、共起回数をエッジの重み（Weight）として扱う)
- **類似した対象同士がどのようなグループを形成しているのか？** → `クラスタ`
- **全体的に関係がどの程度密につながっているのか？** → `ネットワーク密度`

例えば、K-pop関連のニュースであれば、次のように視覚的に確認することができる。

新聞記事のような`非構造化データ`には、`さまざまな対象`と`対象の間の関係`が隠れている。  
`ネットワーク分析`とは、`これらの関係をノードとエッジで表現し`、`どの対象が中心にあるのか`、`どの対象同士がグループを形成しているのか`を探索する方法である。

<br>

#### 例)

1つ目の記事のキーワードを1行に配置し、1つの列に1つのキーワードを入れ、`キーワード同士を互いに接続する。`

1つ目の記事（1行目）のキーワードが `a, b, c, d` の場合、

<div style="background-color:#232627; padding:16px 18px;">
<pre style="background-color:#232627; color:#f2f2f2; margin:0; padding:0;">
a, b, c, d
</pre>
</div>

次のような関係が生成される。

```text
a-b
a-c
a-d
b-c
b-d
c-d
```

**全体をまとめて見ると、**

<div style="background-color:#232627; padding:16px 22px;">
<pre style="background-color:#232627; color:#f2f2f2; margin:0; padding:0; line-height:1.5;">
a ─── c
│╲   ╱│
│ ╲ ╱ │
│ ╱ ╲ │
│╱   ╲│
b ─── d
</pre>
</div>

**2つ目の記事（2行目）のキーワードが `b, d, e, f` の場合、**

<div style="background-color:#232627; padding:16px 22px;">
<pre style="background-color:#232627; color:#f2f2f2; margin:0; padding:0; line-height:1.5;">
b ─── d
│╲   ╱│
│ ╲ ╱ │
│ ╱ ╲ │
│╱   ╲│
e ─── f
</pre>
</div>

**1行目と2行目を接続すると**

<div style="background-color:#232627; padding:16px 22px;">
<pre style="background-color:#232627; color:#f2f2f2; margin:0; padding:0; line-height:1.5;">
a ─── c
│╲   ╱│
│ ╲ ╱ │
│ ╱ ╲ │
│╱   ╲│
b ═══ d
│╲   ╱│
│ ╲ ╱ │
│ ╱ ╲ │
│╱   ╲│
e ─── f
</pre>
</div>

`b-d` は1行目と2行目の両方に登場するため、**エッジの重み（Weight）= 2** となる。
そのため、上の図では `b-d` のエッジを太く表現している。

<span style="color:red;"><b>
1つの記事に10人の人物名が登場し、人物名をノードとして設定した場合、ノードは10個、エッジは最大 (10×9)/2 = 45個となる。
</b></span>

<span style="color:red;"><b>
記事1に千尋、ハクが登場し、
</b></span>

<span style="color:red;"><b>
記事2にも千尋、ハクが登場した場合、
</b></span>

<span style="color:red;"><b>
千尋－ハク間のエッジの重み（Weight）は2となる。
</b></span>

<br>

<span style="color:red;"><b>
ここで `b-d` の接続は1行目と2行目の両方に登場するため、重みが2となる。その他のエッジの重みは1である。
</b></span>

<span style="color:red;"><b>
ネットワーク分析では、ノード（頻度・重要度）だけでなく、エッジ（接続・関係）の重み、つまり2つの単語や人物がどの程度頻繁に一緒に登場したかを視覚的に探索することも目的となる。
</b></span>

| **視覚的要素** | **意味** |
|---|---|
| ノードの大きさ | 単語の頻度、または接続の程度 |
| ノードの色 | クラスタ（Community / Cluster） |
| **エッジの太さ** | `共起回数・エッジの重み（Weight）` |
| エッジの数 | 接続されている単語の数 |
| ノードの位置 | ネットワーク構造上の近さ |
