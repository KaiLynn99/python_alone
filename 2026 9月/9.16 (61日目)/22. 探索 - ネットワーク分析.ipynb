## 2. ネットワークの2つの核心要素

#### 1) `ノード（Node）＝ 対象`

新聞記事の場合、

- 人物
- 企業
- 地域
- 単語
- 事件

などが**ノード**になる。

#### 2) `エッジ（Edge）＝ 関係`

例えば、2つの企業が同じ記事に登場した場合、

## 3. 何が分かるのか？

ネットワークを描くことで、単純な記事内容だけでは見えなかったパターンを見つけることができる。

- **最も多くつながっているのは誰か？** → `中心性`
- **どの単語・人物が互いに強くつながっているのか？** → `共起回数` <br>
(※ 本ノートでは、共起回数をエッジの重み（Weight）として扱う)
- **類似した対象同士がどのようなグループを形成しているのか？** → `クラスタ`
- **全体的に関係がどの程度密につながっているのか？** → `ネットワーク密度`

例えば、K-pop関連のニュースであれば、次のように視覚的に確認することができる。

- 新聞記事のような`非構造化データ`には、`さまざまな対象`と`対象の間の関係`が隠れている。  
- `ネットワーク分析`とは、`これらの関係をノードとエッジで表現し`、`どの対象が中心にあるのか`、`どの対象同士がグループを形成しているのか`を探索する方法である。

<br>

#### 1) Degree（接続数）

各ノード（人物・単語）が、いくつの異なるノードと接続しているかを表す接続数（Degree）を全数調査して計算する。

- <span style="color:red;">右側の `Statistics` タブ → `Average Degree` の横にある `Run` をクリック</span>  
**(→「ノードの大きさ（Size）を接続数（Degree）に比例して大きくする」オプションを適用するため)**

- 左上の `Appearance` → `Nodes` → `Size`（円形アイコン）→ `Ranking`

- `Dropdown`：`Degree` を選択（Min: 20, Max: 200）→ `Apply`

- (記事に最も頻繁に登場し、他の単語と多くつながっている `김민준`、`이서연`、`한지우` などが大きく表示される)

<br>

#### 2) Modularity Class（2つ以上のクラスタに分類）

- <span style="color:red;">右側の `Statistics` タブ → `Modularity` の横にある `Run` をクリック</span>  
(→ Gephiはルーヴァン・アルゴリズム（Louvain Algorithm）を使用し、ネットワーク内のコミュニティ（クラスタ）を自動的に分類する)

- 左上の `Appearance` → `Nodes` → `Color`（パレットアイコン）→ `Partition`

- `Dropdown`：`Modularity Class` を選択 → `Apply`

<br>

#### 3) レイアウトとラベル

- 下部ツールバーの `T` ボタンを押して単語ラベルを表示し、韓国語フォントを選択する。

<br>

#### 4) ノードをマウスで移動して整理

- ノードをマウスで移動し、見やすいように配置を整理する。
