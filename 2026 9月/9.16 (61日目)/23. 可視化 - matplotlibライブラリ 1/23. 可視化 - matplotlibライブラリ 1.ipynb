## 1. Figureを n × m に分割して配置する

#### 使用画像

- `cat.jpg`
- `dog.jpg`
- `flower.jpg`
- `car.jpg`

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg


# ============================================================
# 1. 画像を読み込む
# ============================================================

# 画像ファイルを読み込む
# imread() → 画像ファイルを読み込み、配列形式で返す

img1 = mpimg.imread("cat.jpg")
img2 = mpimg.imread("dog.jpg")
img3 = mpimg.imread("flower.jpg")
img4 = mpimg.imread("car.jpg")


# ============================================================
# 2. Figure全体を作成
# ============================================================

# 複数の画像を配置するための全体の描画領域（Figure）を作成する
#
# figsize=(10, 8)
# → Figure全体のサイズ
# → 横10インチ、縦8インチ

fig = plt.figure(figsize=(10, 8))


# ============================================================
# 3. 1つ目の画像領域を作成
# ============================================================

# Figureを2行2列に分割し、
# その中の1番目の位置にグラフ領域（Axes）を作成する
axes1 = fig.add_subplot(2, 2, 1)

# 1つ目の画像を表示
axes1.imshow(img1)

# 画像のタイトルを設定
axes1.set_title("Cat")

# x軸・y軸の目盛りを非表示にする
axes1.axis("off")


# ============================================================
# 4. 2つ目の画像領域を作成
# ============================================================

# 2行2列のうち2番目の位置
axes2 = fig.add_subplot(2, 2, 2)

# 2つ目の画像を表示
axes2.imshow(img2)

# 画像のタイトルを設定
axes2.set_title("Dog")

# 軸を非表示にする
axes2.axis("off")


# ============================================================
# 5. 3つ目の画像領域を作成
# ============================================================

# 2行2列のうち3番目の位置
axes3 = fig.add_subplot(2, 2, 3)

# 3つ目の画像を表示
axes3.imshow(img3)

# 画像のタイトルを設定
axes3.set_title("Flower")

# 軸を非表示にする
axes3.axis("off")


# ============================================================
# 6. 4つ目の画像領域を作成
# ============================================================

# 2行2列のうち4番目の位置
axes4 = fig.add_subplot(2, 2, 4)

# 4つ目の画像を表示
axes4.imshow(img4)

# 画像のタイトルを設定
axes4.set_title("Car")

# 軸を非表示にする
axes4.axis("off")


# ============================================================
# 7. 画像間の余白を調整
# ============================================================

# 画像やタイトル同士が重ならないように、
# レイアウトを自動的に調整する
fig.tight_layout()


# ============================================================
# 8. Figureを表示
# ============================================================

# 完成したFigureを画面に表示する
plt.show()
```

## 2. アンスコムのデータ

`アンスコムのカルテット（Anscombe's quartet）`は、平均、分散、相関係数、回帰直線がほぼ同じ4つのデータセットであるが、実際のデータ分布やグラフの形はそれぞれ大きく異なる代表的なデータである。

```python
import matplotlib.pyplot as plt

x1 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y1 = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]

x2 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y2 = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]

x3 = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y3 = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]

x4 = [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8]
y4 = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
```

```python
import numpy as np

# 4つのデータセットを1つにまとめる
datasets = [
    (x1, y1),
    (x2, y2),
    (x3, y3),
    (x4, y4)
]


# データセットごとに統計量を計算
for i, (x, y) in enumerate(datasets, start=1):

    # リストをNumPy配列に変換
    x = np.array(x)
    y = np.array(y)

    # 平均
    x_mean = np.mean(x)
    y_mean = np.mean(y)

    # 分散
    # ddof=1 → 標本分散
    x_var = np.var(x, ddof=1)
    y_var = np.var(y, ddof=1)

    # 相関係数
    correlation = np.corrcoef(x, y)[0, 1]

    # 回帰直線
    # y = a*x + b
    # a = 傾き、b = 切片
    a, b = np.polyfit(x, y, 1)

    print(f"\n[Dataset {i}]")

    print(f"x 平均       : {x_mean:.2f}")
    print(f"y 平均       : {y_mean:.2f}")

    print(f"x 分散       : {x_var:.2f}")
    print(f"y 分散       : {y_var:.2f}")

    print(f"相関係数     : {correlation:.2f}")

    print(f"回帰直線の傾き : {a:.2f}")
    print(f"回帰直線の切片 : {b:.2f}")

    print(f"回帰式       : y = {a:.2f}x + {b:.2f}")
```

```python
import matplotlib.pyplot as plt


# ============================================================
# 1. Figure全体を作成
# ============================================================

# グラフ全体を配置するための大きな描画領域（Figure）を作成する
fig = plt.figure(figsize=(10, 8))


# ============================================================
# 2. 1つ目のグラフ領域を作成
# ============================================================

# Figureを2行2列に分割し、その中の1番目の位置に
# グラフ領域（Axes）を作成する
axes1 = fig.add_subplot(2, 2, 1)

# 1つ目のデータセットを散布図で描画する
axes1.scatter(x1, y1)

# グラフのタイトルを設定する
axes1.set_title("Dataset 1")


# ============================================================
# 3. 2つ目のグラフ領域を作成
# ============================================================

# 2行2列のうち2番目の位置に
# グラフ領域（Axes）を作成する
axes2 = fig.add_subplot(2, 2, 2)

# 2つ目のデータセットを散布図で描画する
axes2.scatter(x2, y2)

# グラフのタイトルを設定する
axes2.set_title("Dataset 2")


# ============================================================
# 4. 3つ目のグラフ領域を作成
# ============================================================

# 2行2列のうち3番目の位置に
# グラフ領域（Axes）を作成する
axes3 = fig.add_subplot(2, 2, 3)

# 3つ目のデータセットを散布図で描画する
axes3.scatter(x3, y3)

# グラフのタイトルを設定する
axes3.set_title("Dataset 3")


# ============================================================
# 5. 4つ目のグラフ領域を作成
# ============================================================

# 2行2列のうち4番目の位置に
# グラフ領域（Axes）を作成する
axes4 = fig.add_subplot(2, 2, 4)

# 4つ目のデータセットを散布図で描画する
axes4.scatter(x4, y4)

# グラフのタイトルを設定する
axes4.set_title("Dataset 4")


# ============================================================
# 6. グラフ間の余白を調整
# ============================================================

# 各グラフのタイトルなどが重ならないように
# グラフ間の余白を自動的に調整する
fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

# 完成したFigureを画面に表示する
plt.show()
```
