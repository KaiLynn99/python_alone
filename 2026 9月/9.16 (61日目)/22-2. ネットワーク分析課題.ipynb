#### - 分析結果レポート

<div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>1. データ収集・検索条件</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    当初は、BIG KINDSで「スタジオジブリ」を検索語として設定した。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    しかし、最近の<span style="color:red;"><b>AI、イラスト、画像生成</b></span>に関する記事が多数含まれており、作品を中心とした分析が難しかった。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    そのため、検索語を<b>ジブリアニメーションの作品名</b>を中心としたものに変更し、
    「生成AI」「ChatGPT」「AIイラスト」「画像生成」「人工知能」などを除外語として設定した。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    最終的に、<span style="color:red;"><b>直近約5年間のニュース記事94件</b></span>を分析対象として使用した。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>2. ネットワークの主要ノード（Degree & Size）</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    ノードの大きさは、他のノードとの接続数を表す<span style="color:red;"><b>Degree</b></span>を基準として設定した。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:red;"><b>「千と千尋の神隠し」</b></span>が最も多様なキーワードと接続しており、<b>主要なハブ</b>として現れた。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    その他にも、<b>宮崎駿、スタジオジブリ、となりのトトロ、もののけ姫、天空の城ラピュタ</b>などが比較的大きなノードとして現れた。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>3. Modularity Class（クラスタ / コミュニティ）</b>
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:red;"><b>Modularity分析</b></span>の結果、ネットワークは大きく<b>3つのコミュニティ</b>に分類された。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <span style="color:#7e57c2;"><b>紫色のクラスタ：</b></span>
    千と千尋の神隠し、となりのトトロ、ハウルの動く城、魔女の宅急便など
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:#ef6c00;"><b>オレンジ色のクラスタ：</b></span>
    宮崎駿、天空の城ラピュタ、風の谷のナウシカ、もののけ姫など
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <span style="color:#2e7d32;"><b>緑色のクラスタ：</b></span>
    猫の恩返しを中心として形成された小規模なクラスタ
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    特に<span style="color:#7e57c2;"><b>紫色のクラスタ</b></span>では、
    <b>「公演」「オペラ」「芸術の殿堂（예술의전당）」</b>などのキーワードが一緒に現れた。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>4. 接続強度（Edge Weight）</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    エッジの太さは、2つのキーワードが同じ記事内に一緒に登場した回数を表す
    <span style="color:red;"><b>Weight</b></span>を意味する。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>主な接続は以下のとおりである。</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>千と千尋の神隠し</b> ↔ 公演
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>千と千尋の神隠し</b> ↔ スタジオジブリ
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>千と千尋の神隠し</b> ↔ 宮崎駿
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>宮崎駿</b> ↔ もののけ姫
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>宮崎駿</b> ↔ 天空の城ラピュタ
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    これにより、<span style="color:red;"><b>「千と千尋の神隠し」と宮崎駿がネットワークの中心的な役割を担っている</b></span>ことを確認できた。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>5. 総合結果</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    ジブリ関連ニュースでは、<b>作品、監督、公演、自然、人間</b>などのキーワードが互いに接続し、複数のクラスタを形成していた。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    特に<span style="color:red;"><b>「千と千尋の神隠し」は公演関連のキーワードと強い接続</b></span>を示した。
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    また、<span style="color:red;"><b>宮崎駿は複数のジブリ作品をつなぐ中心ノード</b></span>として現れた。
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    ※ 本結果は、<b>直近約5年間に収集した94件のニュース記事における共起関係</b>を分析したものであり、
    <span style="color:red;"><b>ジブリ作品全体の人気ランキングを意味するものではない。</b></span>
  </div>
