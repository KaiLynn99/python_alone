#### - 분석 결과 보고서

<div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>1. 데이터 수집 및 검색 조건</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    초기에는 BIG KINDS에서 <b>‘스튜디오 지브리’</b>를 검색어로 설정하였다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    그러나 최근 <span style="color:red;"><b>AI, 그림, 이미지 생성</b></span> 관련 기사가 다수 포함되어 작품 중심 분석이 어려웠다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    따라서 검색어를 <b>지브리 애니메이션 작품명</b> 중심으로 변경하고,
    ‘생성형AI’, ‘ChatGPT’, ‘AI그림’, ‘이미지생성’, ‘인공지능’ 등의 단어를 제외어로 설정하였다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    최종적으로 <span style="color:red;"><b>최근 약 5년간 94건의 뉴스 기사</b></span>를 분석 대상으로 사용하였다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>2. 네트워크 핵심 노드 (Degree & Size)</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    노드의 크기는 다른 노드와 연결된 정도를 나타내는 <span style="color:red;"><b>Degree</b></span>를 기준으로 설정하였다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:red;"><b>‘센과 치히로의 행방불명’</b></span>이 가장 다양한 키워드와 연결되어 <b>핵심 허브</b>로 나타났다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    그 외에도 <b>미야자키 하야오, 스튜디오 지브리, 이웃집 토토로, 모노노케 히메, 천공의 성 라퓨타</b> 등이 비교적 큰 노드로 나타났다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>3. Modularity Class (군집 / 커뮤니티)</b>
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:red;"><b>Modularity 분석</b></span> 결과, 네트워크는 크게 <b>3개의 커뮤니티</b>로 구분되었다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <span style="color:#7e57c2;"><b>보라색 군집:</b></span>
    센과 치히로의 행방불명, 이웃집 토토로, 하울의 움직이는 성, 마녀 배달부 키키 등
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <span style="color:#ef6c00;"><b>주황색 군집:</b></span>
    미야자키 하야오, 천공의 성 라퓨타, 바람계곡의 나우시카, 모노노케 히메 등
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <span style="color:#2e7d32;"><b>초록색 군집:</b></span>
    고양이의 보은을 중심으로 형성된 소규모 군집
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    특히 <span style="color:#7e57c2;"><b>보라색 군집</b></span>에서는
    <b>‘공연’, ‘오페라’, ‘예술의전당’</b> 등의 키워드가 함께 나타났다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>4. 연결 강도 (Edge Weight)</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    선의 굵기는 두 키워드가 같은 기사에 함께 등장한 횟수를 나타내는
    <span style="color:red;"><b>Weight</b></span>를 의미한다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>주요 연결은 다음과 같다.</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>센과 치히로의 행방불명</b> ↔ 공연
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>센과 치히로의 행방불명</b> ↔ 스튜디오 지브리
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>센과 치히로의 행방불명</b> ↔ 미야자키 하야오
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>미야자키 하야오</b> ↔ 모노노케 히메
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    <b>미야자키 하야오</b> ↔ 천공의 성 라퓨타
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    이를 통해 <span style="color:red;"><b>센과 치히로의 행방불명과 미야자키 하야오가 네트워크의 중심 역할을 하고 있음</b></span>을 확인할 수 있었다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:10px 14px;">
    &nbsp;
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    <b>5. 종합 결과</b>
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    지브리 관련 뉴스에서는 <b>작품, 감독, 공연, 자연, 인간</b> 등의 키워드가 서로 연결되어 여러 군집을 형성하였다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    특히 <span style="color:red;"><b>센과 치히로의 행방불명은 공연 관련 키워드와 강한 연결</b></span>을 보였다.
  </div>

  <div style="background-color:#f5f5f5; color:#111111; padding:8px 14px;">
    또한 <span style="color:red;"><b>미야자키 하야오는 여러 지브리 작품을 연결하는 중심 노드</b></span>로 나타났다.
  </div>

  <div style="background-color:#e9ecef; color:#111111; padding:8px 14px;">
    ※ 본 결과는 <b>최근 약 5년간 수집한 94건의 뉴스 기사 내 동시출현 관계</b>를 분석한 것이며,
    <span style="color:red;"><b>지브리 작품 전체의 인기 순위를 의미하지 않는다.</b></span>
  </div>
