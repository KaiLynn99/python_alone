- `groupby()`は、特定の列で同じ値を持つデータをグループ化し、  
グループごとに`平均・合計・件数`などを計算するときに使用する。

- SQLの`GROUP BY`とほぼ同じ考え方である。

```python
import pandas as pd


df = pd.DataFrame({
    "部署": ["営業", "営業", "開発", "開発", "人事"],
    "名前": ["千尋", "ハウル", "ソフィー", "アシタカ", "キキ"],
    "売上": [100, 150, 200, 250, 120]
})

display(df)
```

<br>

#### - グループごとの平均を求める

```python
result = df.groupby("部署")["売上"].mean()

display(result)
```

<br>

#### # groupby()で使用する主な集計メソッド

`groupby()`でグループ化した後、`sum()`・`mean()`・`count()`・`max()`・`min()`などを使用して集計できる。

```python
# 部署別の売上合計
display(df.groupby("部署")["売上"].sum())
```

<br>

```python
# 部署別の売上平均
display(df.groupby("部署")["売上"].mean())
```

<br>

```python
# 部署別のデータ件数
display(df.groupby("部署")["売上"].count())
```

<br>

```python
# 部署別の最大売上
display(df.groupby("部署")["売上"].max())
```

<br>

```python
# 部署別の最小売上
display(df.groupby("部署")["売上"].min())
```

<br>

<span style="color:red; font-weight:bold;">
- groupby(), transform()メソッドを利用したカテゴリ別派生変数の追加例)
</span>

```python
import pandas as pd


df = pd.DataFrame({
    "部署": ["営業", "営業", "開発", "開発", "人事"],
    "名前": ["千尋", "ハウル", "ソフィー", "アシタカ", "キキ"],
    "売上": [100, 150, 200, 250, 120]
})

display(df)
```

<br>

`groupby().mean()`はグループごとに`1つの結果`だけを返すが、`transform("mean")`は`元の行数と同じ数の結果`を返す。

```python
# 部署ごとの平均売上を元の行数と同じ形で返す
df.groupby("部署")["売上"].transform("mean")
```

<br>

```python
# 上の結果を新しい派生変数としてDataFrameに追加
df["部署別平均売上"] = df.groupby("部署")["売上"].transform("mean")

display(df)
```
