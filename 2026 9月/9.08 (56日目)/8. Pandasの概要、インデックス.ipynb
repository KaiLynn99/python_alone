#### 1) Seriesのインデックス

Pythonにおける一般的な`インデックス`は、データの`位置`を表す概念である。

一方、Pandasのデータ型では、

- `位置インデックス（Position Index）`
- `ラベルインデックス（Label Index）`

を区別する。

Pandasで単に`インデックス`という場合、基本的には`ラベルインデックス`を意味する。

`ラベルインデックス`を指定しない場合は、自動的に`0, 1, 2, ...`が付与される。

````python
import pandas as pd

s1 = pd.Series([10, 20, 30, 40, 50])
print(s1)
print(type(s1))
````

`0, 1, ...` → `index（インデックス）`  
`10, 20, ...` → `value（値）`

| 位置（position） | インデックス（label） | 値 |
|---:|---:|---:|
| 0 | 0 | 10 |
| 1 | 1 | 20 |
| 2 | 2 | 30 |
| 3 | 3 | 40 |
| 4 | 4 | 50 |

- `インデックス名`を指定してデータを取得 → `loc[]`を使用
- `位置番号`を指定してデータを取得 → `iloc[]`を使用（`integer location`）

```python
# インデックスと位置番号が偶然同じ場合
print(s1.loc[0])    # 10 → インデックス「0」
print(s1.iloc[0])   # 10 → 0番目の位置
```

`Series`は必ず`インデックス`を持つ。

インデックスを指定しない場合、Pandasが自動的に`0, 1, 2, ...`というインデックスを付与する。

この場合、`インデックス`と`位置番号`は同じように見えるが、`概念としては異なる。`

#### - インデックスを直接指定

````python
s2 = pd.Series(
    [90, 80, 70],
    index=["国語", "英語", "数学"]
)

print(s2)
````

| 位置（position） | インデックス（label） | 値 |
|---:|:---:|---:|
| 0 | 国語 | 90 |
| 1 | 英語 | 80 |
| 2 | 数学 | 70 |

```python
print(s2.loc["国語"])    # インデックス名「国語」 → 90
print(s2.iloc[0])       # 0番目の位置 → 90

# 以下のコードはエラー
print(s2.loc[0])
```

-> `loc[]`は`インデックス（ラベル）`を基準にデータを取得するため、  
インデックスが`"国語"`、`"英語"`、`"数学"`の場合、`s2.loc[0]`では取得できない。

#### 2) DataFrameのインデックス

````python
import pandas as pd

df1 = pd.DataFrame({
    "名前": ["千尋", "ハウル", "ソフィー"],
    "年齢": [20, 30, 25],
    "点数": [90, 80, 95]
})

print(df1)
````

````python
# 行の位置番号と行インデックスが同じ
print(df1.loc[0])
print("---")
print(df1.iloc[0])
````

````python
import pandas as pd

df2 = pd.DataFrame({
    "名前": ["千尋", "ハウル", "ソフィー"],
    "年齢": [20, 30, 25],
    "点数": [90, 80, 95]
}, index=["a", "b", "c"])

print(df2)
````

````python
print(df2.loc["a"])
print("---")
print(df1.iloc[0])
````

- データ分析のコードでは、ラベルインデックスのメソッドである`loc[]`はあまり使用されない。  
- ほとんどの場合、位置を基準とするメソッドである`iloc[]`を使用する。
