## Series

- 同一のデータ型を持つ複数の要素で構成されるデータ構造
- NumPy配列と同様の`1次元array`
- `値（values）`と`インデックス（index）`で1次元に構成される
- `Series`を作成する際に`index`を指定しない場合、位置番号と同じ`0`から始まる整数型の基本インデックスが付与される
- 基本インデックスの代わりに、ユーザーが直接指定することもできる

```python
s = pd.Series([3, -5, 7, 4], index=["A", "B", "C", "D"])
```

````python
import pandas as pd

s = pd.Series([3, -5, 7, 4], index=["A", "B", "C", "D"])

print(s)
````

## 1. Seriesの情報確認

`Series`型は、`pandas`モジュール内の`Series`クラスのコンストラクタ関数を使用する。

```python
pandas.Series(data=None, index=None, dtype=None, name=None, copy=None)
```

- `data`引数：`array-like`、`Iterable`、`dict`、または`scalar value`

<img src="../images/image-4.png" width="600">

````python
import pandas as pd

s = pd.Series(
    data=[90, 80, 70],
    index=["千尋", "ハウル", "キキ"],
    dtype="int64",
    name="点数"
)

print(s)
````

````python
import pandas as pd

# Seriesを作成
temperature = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

print("INDEX: ", temperature.index)
print("VALUES: ", temperature.values)
print(temperature)
````

````python
temperature.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
print("INDEX: ", temperature.index)
print("VALUES: ", temperature.values)
print(temperature)
````

````python
# Seriesの情報を確認
print("データ型: ", temperature.dtype)
print("データ数: ", temperature.size)
````

````python
# info()はデータ数、データ型、メモリ使用量などの情報を確認するときに使用する
print(temperature.info())
````

````python
print("shape: ", temperature.shape)  # (12,) → 1次元のデータが12個
print("dtype: ", temperature.dtype)
````

## 2. スライシングと算術演算

````python
# loc[]：インデックス基準
# vs
# iloc[]：位置（番号）基準
print(temperature.loc["Jan":"Mar"])  # Marまで含む → Jan, Feb, Mar

print()

print(temperature.iloc[0:2])         # 2は含まれない → 0, 1
````

\- データ前処理の過程では、`位置基準`の作業が必要な場合が多い。`iloc[]`メソッドを使用する方が便利である。<br>
\- 特別な場合、位置ではなく`意味のある名前`でデータを選択する必要がある場合にのみ`loc[]`を使用する。

次の場合は`iloc[]`が便利である。

- 最初／最後のデータを選択
- 特定の位置のデータを選択
- データサンプルを抽出
- 反復的な位置基準の処理
- インデックス名が変更される可能性があるデータ

````python
# スライシング
print(temperature.iloc[:3])    # 先頭から3個（位置0～2）
print(temperature.iloc[-3:])   # 末尾から3個
print(temperature.iloc[1:5])   # 位置1～4（2番目～5番目）
print(temperature.iloc[::2])   # 位置0, 2, 4, ...を選択
print(temperature.iloc[::-1])  # 逆順に並べる
````

````python
# 条件抽出
print(temperature[temperature >= 20])  # 20以上の値を選択
print(temperature[temperature < 0])    # 0未満の値を選択
````

````python
# 算術演算
print(temperature + 1)              # すべての値に1を加算
print(temperature * 2)              # すべての値を2倍
print(temperature - 5)              # すべての値から5を減算
print(temperature / 2)              # すべての値を2で除算
print(temperature.iloc[:3] + 10)    # 先頭3個の値に10を加算
````
