## 1. ヒストグラムと密度曲線を同時に可視化

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Yu Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
# ============================================================
# 1. データ準備
# ============================================================

# Seabornが提供するtipsデータセットを読み込む。
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置する大きなキャンバス（Figure）を作成する。
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域（Axes）を作成する。
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. ヒストグラム + 密度曲線を描画
# ============================================================

# total_billの分布をヒストグラムで表現する。
#
# histplot()
# → ヒストグラムを描画するSeabornの関数
#
# kde=True
# → KDE（Kernel Density Estimation）
#   つまり、密度曲線も一緒に表示する。
#
# bins=10
# → ヒストグラムの区間を10個に分ける。

sns.histplot(
    data=tips,
    x="total_bill",
    bins=10,
    kde=True,
    ax=axes1
)


# ============================================================
# 5. グラフ設定
# ============================================================

# グラフタイトル
axes1.set_title("合計金額の分布")

# x軸名
axes1.set_xlabel("合計金額")

# y軸名
axes1.set_ylabel("頻度")


# ============================================================
# 6. グラフの間隔を調整
# ============================================================

fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

plt.show()
```

## 2. 散布図に回帰線を追加

```python
# ============================================================
# 1. データ準備
# ============================================================

# Seabornが提供するtipsデータセットを読み込む。
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())


# ============================================================
# 2. Figure全体を作成
# ============================================================

# グラフ全体を配置する描画領域（Figure）を作成する。
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. グラフ領域（Axes）を作成
# ============================================================

# 1行1列のうち1番目の位置に
# グラフ領域（Axes）を作成する。
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 散布図 + 回帰線を描画
# ============================================================

# total_billとtipの関係を散布図で表示し、
# 2つの変数の線形的な関係を表す回帰線も一緒に描画する。
#
# data=tips
# → 使用するDataFrame
#
# x="total_bill"
# → x軸に使用する列
#
# y="tip"
# → y軸に使用する列
#
# ax=axes1
# → 作成したaxes1領域にグラフを描画する。
#
# ci=None
# → 回帰線の周囲に信頼区間を表示しない。
# → 入門例では回帰線のみを表示するために使用する。

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=axes1,
    ci=None
)


# ============================================================
# 5. グラフ設定
# ============================================================

# グラフタイトル
axes1.set_title("合計金額とチップの関係")

# x軸名
axes1.set_xlabel("合計金額")

# y軸名
axes1.set_ylabel("チップ")


# ============================================================
# 6. Figureの間隔を調整
# ============================================================

fig.tight_layout()


# ============================================================
# 7. Figureを表示
# ============================================================

plt.show()
```

## 3. Seabornの`jointplot()`で2つの数値型変数の関係 + それぞれの分布を同時に確認

```python
# ============================================================
# 1. データ準備
# ============================================================

# Seabornが提供するtipsデータセットを読み込む。
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())


# ============================================================
# 2. 散布図 + ヒストグラム
# ============================================================

# jointplot()
# → 2つの変数の関係と、それぞれの変数の分布を同時に表示する。
#
# 中央   → 散布図
# 上部   → x軸変数のヒストグラム
# 右側   → y軸変数のヒストグラム
#
# x="total_bill"
# → x軸変数
#
# y="tip"
# → y軸変数

sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip"
)


# ============================================================
# 3. グラフを表示
# ============================================================

plt.show()
```

## 4. バイオリンプロット（Violin Plot）は、見た目はボックスプロットと似ている。

(しかし、単純に四角い箱を表示するだけではなく、データの分布の形もあわせて確認できるグラフ)

```python
# tipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())

# Figureを作成
fig = plt.figure(figsize=(8, 5))

# Figure内に1つのAxesを作成
axes1 = fig.add_subplot(1, 1, 1)

# tipの分布をバイオリン形状で表現
sns.violinplot(
    data=tips,
    y="tip",
    ax=axes1
)

# タイトルと軸名を設定
axes1.set_title("チップのバイオリンプロット")
axes1.set_ylabel("チップ")

# グラフの余白を調整
fig.tight_layout()

# グラフを表示
plt.show()
```

`countplot()`は`value_counts() + bar()`を便利にまとめた機能である。  
`countplot()`では、個数を別途計算する必要がない。

```python
# tipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())

# 性別ごとのデータ数を確認
display(tips["sex"].value_counts())

# Figureを作成
fig = plt.figure(figsize=(8, 5))

# Figure内に1つのAxesを作成
axes1 = fig.add_subplot(1, 1, 1)


# Matplotlibにはcountplot()という関数がない。
# そのため、カテゴリ別の個数を先に計算してからbar()で描画する必要がある。

# 性別ごとのデータ数を棒グラフで表現
sns.countplot(
    data=tips,
    x="sex",
    ax=axes1
)

# タイトルと軸名を設定
axes1.set_title("性別ごとの顧客数")
axes1.set_xlabel("性別")
axes1.set_ylabel("件数")

# グラフの余白を調整
fig.tight_layout()

# グラフを表示
plt.show()
```

## 6. `pairplot()`は複数の数値型変数間の関係を一度に探索するグラフ

<span style="color:#1e88e5;">pandasにも同じメソッドがある</span>

※ `pairplot()`はデータセット探索の初期段階で、EDAの探索用として使用

- 変数間の相関関係
- 正の関係 / 負の関係
- 変数の分布
- 異常に突出したデータ
- カテゴリ別の分布の違い

```python
import matplotlib.pyplot as plt
import seaborn as sns


# tipsデータセットを読み込む
tips = sns.load_dataset("tips")

# データ確認
display(tips.head())


# 複数の変数間の関係を一度に可視化
sns.pairplot(
    data=tips
)


# グラフを表示
plt.show()
```

## 7. `heatmap()`は数値データを色の濃淡で表現するグラフ

(特にSeabornでは、`相関係数行列を可視化する際によく使用する`)

```python
# mpgデータセットを読み込む
mpg = sns.load_dataset("mpg")

# データ確認
display(mpg.head())


# 数値型変数の相関係数を計算
corr = mpg.corr(numeric_only=True)


# 相関係数行列を確認
display(corr)


# Figureを作成
fig = plt.figure(figsize=(10, 8))


# Figure内に1つのAxesを作成
axes1 = fig.add_subplot(1, 1, 1)


# 相関係数行列をHeatmapで表現
sns.heatmap(
    corr,
    annot=True,      # 各セルに相関係数の数値を表示
    fmt=".2f",       # 小数点以下2桁まで表示
    ax=axes1
)


# タイトルを設定
axes1.set_title("MPGデータの相関ヒートマップ")


# グラフの余白を調整
fig.tight_layout()


# グラフを表示
plt.show()
```
