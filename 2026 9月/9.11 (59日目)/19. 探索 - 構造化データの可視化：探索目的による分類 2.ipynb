```python
import pandas as pd

data = {
    "名前": [f"学生{i}" for i in range(1, 31)],

    # 勉強時間が長いほど試験点数が高くなる関係
    "勉強時間": [
        1, 1, 2, 2, 2, 3, 3, 3, 4, 4,
        4, 5, 5, 5, 6, 6, 6, 7, 7, 7,
        8, 8, 8, 9, 9, 9, 10, 10, 10, 11
    ],

    # 勉強時間が長いほどゲーム時間が短くなる関係
    "ゲーム時間": [
        5.5, 5.0, 5.2, 4.7, 4.5, 4.3, 4.0, 3.8, 3.7, 3.5,
        3.3, 3.0, 2.9, 2.7, 2.5, 2.3, 2.2, 2.0, 1.8, 1.7,
        1.5, 1.4, 1.2, 1.1, 1.0, 0.9, 0.7, 0.6, 0.5, 0.3
    ],

    # 勉強時間と特に関係がないように設定した変数
    "身長": [
        158, 172, 165, 180, 163, 175, 169, 161, 178, 166,
        182, 159, 171, 168, 176, 164, 181, 160, 173, 167,
        179, 162, 170, 177, 165, 183, 169, 174, 161, 178
    ],

    # 勉強時間が増えるほど、全体的に点数も高くなる
    "試験点数": [
        52, 55, 58, 60, 61, 64, 66, 65, 69, 71,
        73, 74, 76, 78, 79, 81, 80, 84, 86, 85,
        87, 89, 88, 91, 92, 90, 94, 95, 96, 98
    ]
}

df = pd.DataFrame(data)

df
```

#### **1) 正の相関関係**

`勉強時間` ↔ `試験点数`

勉強時間が増えるほど、試験点数も高くなる傾向がある。

```python
df.plot(
    kind="scatter",
    x="勉強時間",
    y="試験点数"
);
```

#### **2) 負の相関関係**

`ゲーム時間` ↔ `試験点数`

ゲーム時間が増えるほど、試験点数が低くなる傾向がある。

```python
df.plot(
    kind="scatter",
    x="ゲーム時間",
    y="試験点数"
);
```

#### **3) 関係がない場合**

`勉強時間` ↔ `身長`

勉強時間と身長の間には、明確な線形関係が見られない。

```python
df.plot(
    kind="scatter",
    x="勉強時間",
    y="身長"
);
```

- **ここで「相関関係がある ＝ 原因と結果がある」という意味ではない。**

- **そして、散布図で関係を確認した後、統計量を用いて関係を数値化して確認する。**

- **このデータでは、散布図 → 相関係数（corr()）→ 正・負・無相関を解釈する。**

```python
print("勉強時間 ↔ 試験点数:", df["勉強時間"].corr(df["試験点数"]))
print("ゲーム時間 ↔ 試験点数:", df["ゲーム時間"].corr(df["試験点数"]))
print("勉強時間 ↔ 身長:", df["勉強時間"].corr(df["身長"]))
```

#### ※ corr()の意味

`corr()`は、デフォルトで**ピアソンの相関係数（Pearson correlation coefficient）**を計算する。
