**※ グラフを描画するために、元のデータセットを修正する必要がある場合が多い。**

````python
# すでにカテゴリごとの人数が集計されているデータ
df = pd.Series([7, 10], index=["M", "F"])
df.plot(kind="bar")
````

````python
# value_counts()で性別ごとの人数を集計してから棒グラフを描画
gender.value_counts().plot(kind="bar")
````

#### - 数値型変数を1つ探索

````python
age_series = df["年齢"]
age_series
````

````python
age_series.plot(kind="hist", bins=8)

# binsオプション（棒の数）を変えて探索
# age_series.plot(kind="hist", bins=5)
# age_series.plot(kind="hist", bins=10)
````

##### 解釈：
- 年齢がどの区間に多く分布しているかを確認する
- およそ25歳～35歳の区間に多く分布している
- 75歳は他の値から離れている

````python
age_series.plot(kind="box");

# ボックス = 中央50%のデータが集まっている範囲
# （ボックスが大きいほど中央50%のデータが広く分布し、
#   ボックスが小さいほど比較的まとまっていると解釈する）
# ボックス内の線 = 中央値
# ひげ = 外れ値を除いたデータの範囲
# 点 = 外れ値
````

##### 解釈：
- 8歳と75歳が外れ値として表示される可能性が高い
