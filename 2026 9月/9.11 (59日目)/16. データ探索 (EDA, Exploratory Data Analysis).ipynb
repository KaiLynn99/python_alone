## 1. データ探索（EDA, Exploratory Data Analysis）

データを本格的に分析したりモデリングしたりする前に、`データの特徴、パターン、関係、異常値などを把握する過程`である。

<br>

データ収集  
↓  
データ前処理  
↓  
★ `データ探索（EDA）` 
↓  
可視化  
↓  
統計分析  
↓  
モデリング  
↓  
解釈および意思決定  

#### - データ探索の方法

`数値による探索 vs 可視化による探索`

データ探索
│
├─ 統計による探索
│  ├─ 平均
│  ├─ 中央値
│  ├─ 最頻値
│  ├─ 標準偏差
│  ├─ 最小値 / 最大値
│  └─ 相関関係など
│
└─ 可視化による探索
   ├─ 表
   ├─ グラフ
   ├─ 地図
   └─ テキストの可視化

## 2. データ形式によるデータ探索方法の比較

1) 構造化データ → 統計・チャート

2) 非構造化データ → テキスト・画像・音声・動画

3) 関係データ → ネットワーク分析（Gephi）

## - 構造化データ vs 非構造化データ

### [構造化データ]

| 区分 | 主な問い | 変数 | 代表的なチャート |
| --- | --- | --- | --- |
| 単変量 | 1つの変数にはどのような特徴があるか？ | 1個 | 棒グラフ、ヒストグラム、Boxplot |
| 比較 | 何がより大きいか？ | 1個以上 | 棒グラフ、グループ化棒グラフ、積み上げ棒グラフ |
| 分布 | データはどのように分布しているか？ | 1個以上 | ヒストグラム、Boxplot、KDE |
| 関係 | 変数同士にはどのような関係があるか？ | 2個以上 | 散布図、バブルチャート、ヒートマップ |
| 時間 | 時間の経過に伴ってどのように変化するか？ | 時間 + 値 | 折れ線グラフ、面グラフ |
| 空間 | 地域によってどのように異なるか？ | 位置 + 値 | 地図、バブルマップ、コロプレスマップ |

### [非構造化データ]

| データ | 主な探索対象 | 代表的な可視化 |
| --- | --- | --- |
| テキスト | 単語、頻度、感情、関係、トピック | Word Cloud、棒グラフ、ネットワーク |
| 画像 | 色、オブジェクト、位置、特徴 | 画像、ヒートマップ、Bounding Box |
| 音声 | 波形、周波数、音声内容 | Waveform、Spectrogram |
| 動画 | オブジェクト、位置、時間、行動 | フレーム、Tracking、時系列グラフ |
| Web | テキスト、リンク、画像、構造 | ネットワーク、頻度、時系列 |
