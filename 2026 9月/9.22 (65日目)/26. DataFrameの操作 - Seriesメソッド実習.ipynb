DataFrame = 複数のSeriesが集まった2次元データ構造  
Series = DataFrameから1つの列を選択した1次元データ構造

<span style="color:red;">df["列名"] → Seriesを返す</span>  
<span style="color:red;">df[["列名"]] → DataFrameを返す</span>

````python
import pandas as pd

df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル"],
    "点数": [80, 90, 75]
})
````

````python
# 1列を選択 → Series
s = df["点数"]

display(s)
display(type(s))
````

````python
# 1列をリスト形式で選択 → DataFrame
df2 = df[["点数"]]

display(df2)
display(type(df2))
````

````python
# apply()メソッドに関数を引数として渡す（高階関数）
def sales_grade(value):
    # valueはSeriesではないため、value.isna()は使用できない
    # pd.isna(値)を使用する
    if pd.isna(value):
        return "未集計"
    elif value >= 1000:
        return "優秀"
    elif value >= 800:
        return "普通"
    else:
        return "低い"


# Seriesの各値にsales_grade関数を順番に適用
grade = df["売上"].apply(sales_grade)
display(grade)
````
