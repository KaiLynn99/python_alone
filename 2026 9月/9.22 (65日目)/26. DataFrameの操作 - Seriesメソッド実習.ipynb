DataFrame = 複数のSeriesが集まった2次元データ構造  
Series = DataFrameから1つの列を選択した1次元データ構造

<span style="color:red;">df["列名"] → Seriesを返す</span>  
<span style="color:red;">df[["列名"]] → DataFrameを返す</span>

````python
import pandas as pd

df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル"],
    "点数": [80, 90, 75]
})


# 1列を選択 → Series
s = df["点数"]

display(s)
display(type(s))

print()

# 1列をリスト形式で選択 → DataFrame
df2 = df[["点数"]]

display(df2)
display(type(df2))
````

#### 2) 統計

````python
display("平均:", df["売上"].mean())
display("中央値:", df["売上"].median())

# カテゴリ型 - 最頻値は頻度が同じ場合、
# 2つ以上を返す必要があるためSeries型で返す
display("最頻値:", df["等級"].mode())
display("最頻値:", df["性別"].mode())


display("合計:", df["売上"].sum())
display("最大値:", df["売上"].max())
display("最小値:", df["売上"].min())


display("分散:", df["売上"].var())
display("標準偏差:", df["売上"].std())


display("第1四分位数:", df["売上"].quantile(0.25))
display("第2四分位数:", df["売上"].quantile(0.50))
display("第3四分位数:", df["売上"].quantile(0.75))
````
