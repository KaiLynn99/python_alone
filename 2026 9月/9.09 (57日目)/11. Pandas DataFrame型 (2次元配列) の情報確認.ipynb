- 表のようなスプレッドシート形式のデータ構造

- `2次元array`は同一のデータ型を持つ必要があるが、`DataFrame`は列ごとに異なるデータ型を持つことができる

## 2. DataFrame情報の確認

<img src="../images/image-10.png" width="1000">

````python
import pandas as pd

df = pd.DataFrame(
    [[90, 80],
     [70, 60]],
    columns=['国語', '英語']
)

print(df)
````

````python
import pandas as pd

# 名前、年齢、性別データのDataFrameを作成
df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー", "アシタカ"],
    "年齢": [20, 21, 20, 22, 21],
    "性別": ["男", "女", "男", "女", "男"]
})

print(df)
````

#### 1) `info()`メソッド

`行数`、`列数`、`列名`、`欠損値ではないデータ数`、`データ型`を確認する。

````python
print("===== info =====")

# DataFrame全体の情報を出力
df.info()
````

#### 2) `shape`, `size`, `ndim`

`行`と`列`のサイズを確認する。

````python
print("shape:", df.shape)  # 出力（行数, 列数）

print("行数:", df.shape[0])
print("列数:", df.shape[1])

print(df.shape, df.size, df.ndim)
````

#### 3) `dtypes`

各列の`データ型`を確認する。

````python
print("===== dtypes =====")
print(df.dtypes)

# 特定の列のデータ型を確認
print(df["年齢"].dtype)
print(df["名前"].dtype)
````

#### 4) `columns`

`列名`を確認する。

````python
print("===== columns =====")
print(df.columns)

# 列名のみ確認
print(df.columns.tolist())

# in演算子を使用して特定の列が存在するか確認
print("年齢" in df.columns)
print("住所" in df.columns)
````

#### 5) `index`

`行ラベルのインデックス`を確認する  
※ `位置インデックス`ではない

````python
print("===== index =====")
print(df.index)
````

#### 6) `head`, `tail`

DataFrameの`先頭`または`末尾`のデータを確認するときに使用する。

````python
print("===== head =====")
print(df.head())   # デフォルトでは5行
print(df.head(3))

print()

print("===== tail =====")
print(df.tail())
print(df.tail(3))
````

#### 7) `unique`, `nunique`

`ユニーク値`の確認、`ユニーク値の個数`を確認する

````python
print(df["性別"].unique())
print(df["年齢"].unique())

print()

# ユニーク値の個数
print(df["性別"].nunique())
````
