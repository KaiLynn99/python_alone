#### 問題 1. `[]`で列を選択

`mpg` DataFrameから`mpg`列のみを選択して出力しなさい。

````python
mpg["mpg"]
````

#### 問題 2. `[]`で複数の列を選択

`mpg` DataFrameから`mpg`、`weight`、`origin`の3つの列を選択して出力しなさい。

````python
mpg[["mpg", "weight", "origin"]]
````

#### 問題 3. `iloc`で特定の行を選択

`mpg`から`10番目の行`のデータを出力しなさい。  
ただし、数値位置を使用する`iloc`を使いなさい。

````python
mpg.iloc[9]
````

#### 問題 4. `iloc`で行 + 列を選択

`mpg`から`3番目の行`の`1番目の列`の値を出力しなさい。

````python
mpg.iloc[2,0]
````

#### 問題 5. `iloc`で行 + 列をスライス

`mpg`から`1～5番目の行`と`1～3番目の列`を選択して出力しなさい。

````python
mpg.iloc[[0,5], [0,3]]
````

#### 問題 6. `loc`で特定の行 + 列を選択

`mpg`から`インデックスが10の行`の`mpg`の値を出力しなさい。

````python
mpg.loc[10, "mpg"]
````

#### 問題 7. `loc`で複数の行 ＋ 列を選択


`インデックスが10〜15の行`で

- `mpg`

- `weight`

- `origin`

`列だけ`を選択して出力しなさい。

````python
mpg.loc[10:15, ["mpg", "weight", "origin"]]
````

#### 問題 8. `loc` スライシング

`mpg` において、`インデックス 10 から 15 まで`のデータを選択しなさい。

````python
mpg.loc[10:15]
````

#### 問題9. `条件`で`行を選択`

`mpg`で、`mpg`が`30以上`の自動車のみを選択しなさい。

````python
mpg.loc[mpg["mpg"] >= 30]
````

#### 問題 10. `条件` + `列の選択`

`mpg`が`30以上`の自動車の`name`と`mpg`のみを出力しなさい。

````python
mpg.loc[mpg["mpg"] >= 30, ["name", "mpg"]]
````

#### 問題 11. `2つの条件`で`行を選択`する

`mpg`が`30以上`で、`weight`が`2000以下`の自動車のみを出力しなさい。

````python
mpg.loc[(mpg["mpg"] >= 30) & (mpg["weight"] <= 2000)]
````

#### 問題12. 次の`コードの意味`を`説明`しなさい。

```py
df.loc[10:15, ['mpg', 'weight']]

df.iloc[10:15, 0:2]

df['mpg']

df[10:15]

df[df['mpg'] >= 30]
```

模範解答

```py
# df.loc[10:15, ['mpg', 'weight']]
# インデックス名が 10 から 15 までの行における、mpg 列と weight 列のデータを取得する。（末尾 15 を含む）

# df.iloc[10:15, 0:2]
# 11〜15 番目の行（インデックス 10〜14）と、1〜2 番目の列（インデックス 0〜1）のデータを取得する。（末尾 15, 2 を含まない）

# df['mpg']
# mpg 列の全データを取得する。

# df[10:15]
# インデックス 10 から 14 までの行（11〜15 番目の行）を取得する。（末尾 15 を含まない）

# df[df['mpg'] >= 30]
# mpg 列の値が 30 以上である行のみを抽出して取得する。
```
