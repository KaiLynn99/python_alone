```python
# opinion列の文章を空白でつなげて1つのテキストにまとめる
text = " ".join(df["opinion"])

# 結合したテキストを確認
print(text)
```

#### - まず数値として探索

```python
from collections import Counter

# すべての文字を小文字に変換し、単語ごとに分割
words = text.lower().split()

# 分割した単語を確認
print(words)

# 単語ごとの出現回数を集計
word_count = Counter(words)

# 単語ごとの出現回数を確認
word_count
```

#### - 可視化

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=set()  # デフォルトのストップワードを使用しない
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

#### - 分析上あまり意味のない単語（ストップワード）を除外

```python
stopwords = {
    "the", "and", "is", "to", "was",
    "a", "of", "in", "with", "me",
    "my", "more", "than", "I"
}

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=stopwords  # ストップワードを設定
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
