```python
# opinion列の文章を空白でつなげて1つのテキストにまとめる
text = " ".join(df["opinion"])

# 結合したテキストを確認
print(text)
```

#### - まず数値として探索

```python
from collections import Counter

# すべての文字を小文字に変換し、単語ごとに分割
words = text.lower().split()

# 分割した単語を確認
print(words)

# 単語ごとの出現回数を集計
word_count = Counter(words)

# 単語ごとの出現回数を確認
word_count
```

#### - 可視化

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=set()  # デフォルトのストップワードを使用しない
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

#### - 分析上あまり意味のない単語（ストップワード）を除外

```python
stopwords = {
    "the", "and", "is", "to", "was",
    "a", "of", "in", "with", "me",
    "my", "more", "than", "I"
}

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=stopwords  # ストップワードを設定
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from collections import Counter
from wordcloud import WordCloud
```

```python
# Excelファイルを読み込む
df = pd.read_excel("news_2.xlsx")

df.head()
```

```python
# 「キーワード」列を抽出
keywords = df["키워드"] # df.loc[], df.iloc[], df[]

keywords.head()
```

```python
# 各行のキーワードを1つの文字列に結合
text = ",".join(keywords)
```

```python
# 全体の文字数を確認
print(len(text))

# 先頭500文字を確認
print(text[:500])
```

```python
# 「,」を基準にキーワードを分割
words = text.split(",")

# キーワードの総数を確認
print(len(words))

# 先頭20個のキーワードを確認
print(words[:20])
```

```python
# 各キーワードの出現回数を集計
word_count = Counter(words)

# 出現回数が多い上位10個を確認
print(word_count.most_common(10))
```

```python
wordcloud2 = WordCloud(
    font_path="C:/Windows/Fonts/malgun.ttf",  # WordCloud内で韓国語が文字化けしないように韓国語フォントを設定
    width=800,
    height=500,
    background_color="white",
    stopwords=set()
).generate_from_frequencies(word_count)

# 出力
plt.figure(figsize=(10, 6))
plt.imshow(wordcloud2)
plt.axis("off")
plt.show()
```

> **頻度を直接計算せず、WordCloudの仕組みを利用することもできる。** <br>
> `generate_from_frequencies()` メソッドの代わりに、 `generate()` メソッドを使用することもできる。

| 区分 | **generate_from_frequencies()** | **generate()** |
|---|---|---|
| 入力 | `頻度辞書` | `文字列 text` |
| 頻度計算 | ユーザーが事前に計算 | `WordCloudが直接行う` |
| 例 | `generate_from_frequencies(words_count)` | `generate(text)` |
| Counterの使用 | 使用可能 | 不要 |
| 頻度の制御 | 正確に制御可能 | `相対的に難しい` |
| 現在のデータ | より推奨 | 使用可能 |

<br>
> ※ `words_count.most_common(20)` の頻度とWordCloudで使用される頻度を同じように維持したい場合は、`generate_from_frequencies()` を使用する。

```python
# WordCloudを生成
wordcloud = WordCloud(
    font_path="C:/Windows/Fonts/malgun.ttf",
    width=800,
    height=500,
    background_color="white"
).generate(text)

# 出力
plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
