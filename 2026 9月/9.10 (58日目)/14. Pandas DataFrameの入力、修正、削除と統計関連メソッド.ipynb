## 1. 入力、修正、削除

````python
import pandas as pd

df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー", "アシタカ"],
    "学年": [1, 2, 1, 3, 2],
    "点数": [85, 92, 78, 95, 88],
    "出席": [90, 95, 80, 98, 85]
})

df
````

#### # 入力

#### 新しい列の追加

`新しい列変数 = Series`：データセットの要素数と`Series`の要素数が同じでなければならない。異なる場合は`error`

`新しい列変数 = list`：`list`を`Series`に自動的に型変換した後、代入される

`新しい列変数 = scalar値`：データセットの要素数と同じ`Series`（同じ値を持つ）に変換した後、代入される

````python
df["成績"] = ["B", "A", "C", "A+", "B+"]

df
````

````python
df["生年"] = 2000  # 1つの値 → 同じ値を持つSeriesに変換した後、代入

df
````

#### 計算して新しい列を作成
`df["点数"] + df["出席"]`の結果は、（`Series + Series`）なので`Series`である。

````python
df["合計点"] = df["点数"] + df["出席"]  # 計算されたSeriesの値が代入される

df
````

#### # 修整

- 特定の値を修正

````python
df.loc[0, "点数"] = 90

df
````

- 特定範囲の値を修正

````python
df.loc[df["点数"] < 80, "点数"] = 80

df
````

##### * 原理

`df.loc[df["点数"] > 90, "点数"] = 99`  
→ 条件式の引数の位置には`Boolean Series型`が入る

`df.loc[pd.Series([False, True, False, True, False]), "点数"] = 99`

`df.loc[[False, True, False, True, False], "点数"] = 99`  
→ `リスト`が`Series`に自動で型変換される

#### `[]`演算子を使用すると列全体が修正される

````python
df["合計点"] = 0

df
````

#### # 削除 

```py
- 行の削除：`drop("行ラベル", axis=0)`
```

- `axis=0`は行軸（縦方向 → 1つ以上の行を削除可能）

```py
- 列の削除：`drop("列ラベル", axis=1)`
```

- `axis=1`は列軸（横方向 → 1つ以上の列を削除可能）

````python
# 「合計点」を列方向（axis=1）から削除

df.drop("合計点", axis=1)  # axisを省略すると0
````

※ `drop()`の引数に`位置の値`を使用することはできない!

※ `特定位置の値`を削除することはできない → 値を空白、NaNなどの`欠損値に変更`することは可能

## 2. 統計関連メソッド

#### 1) 全体の統計量

| 統計量 | 意味 |
|---|---|
| `count` | データ数 |
| `mean` | 平均 |
| `std` | 標準偏差 |
| `min` | 最小値 |
| `25%` | 第1四分位数 |
| `50%` | 中央値 |
| `75%` | 第3四分位数 |
| `max` | 最大値 |

````python
df = pd.DataFrame({
    "名前": ["千尋", "キキ", "ハウル", "ソフィー", "アシタカ"],
    "学年": [1, 2, 1, 3, 2],
    "点数": [85, 92, 78, 95, 88],
    "出席": [90, 95, 80, 98, 85]
})

df
````

````python
# 各列の欠損値を除いたデータ数を返す
print(df.count())
````

````python
# 数値型の列のみ合計を返す
print(df.sum(numeric_only=True))
````

````python
# 数値型の列のみ平均値を返す
print(df.mean(numeric_only=True))
````

````python
# 数値型の列のみ中央値を返す
print(df.median(numeric_only=True))
````

````python
# 数値型の列のみ最小値を返す
print(df.min(numeric_only=True))
````

````python
# 数値型の列のみ最大値を返す
print(df.max(numeric_only=True))
````

````python
# 数値型の列のみ標準偏差を返す
print(df.std(numeric_only=True))
````

````python
# 「点数」列の最大値と最小値の差を返す
print(df["点数"].max() - df["点数"].min())
````

#### 数値型変数の統計量

````python
# 数値型の列の基本統計量をまとめて表示する
df.describe()
````

`学年`の場合、カテゴリ型変数であるにもかかわらず`数値型変数`として設定されているため、統計量が出力される。

`学年`を`カテゴリ型`に変更すると、統計量には出力されない。

````python
# 「学年」のデータ型をcategory型（カテゴリ型）に変換した結果を
# 再びdf["学年"]に保存し、dfの「学年」のデータ型を変更
df["学年"] = df["学年"].astype("category")
````
