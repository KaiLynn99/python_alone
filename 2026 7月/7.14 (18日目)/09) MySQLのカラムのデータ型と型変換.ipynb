## 1. データ型

#### 実務でよく使用する主なデータ型

| 分類 | データ型 | 説明 | 例 |
|---|---|---|---|
| 数値 | `TINYINT` | 非常に小さい整数（-128～127） | 年齢、状態値 |
| 数値 | `INT` | 一般的な整数 | 会員番号、価格 |
| 数値 | `BIGINT` | 非常に大きい整数 | 取引番号、ログ番号 |
| 数値 | `DECIMAL(M, D)` | 正確な小数 | 金額、評価点 |
| 文字列 | `VARCHAR(n)` | 可変長の文字列 | 名前、メールアドレス |
| 文字列 | `TEXT` | 長い文字列 | 投稿内容 |
| 日付 | `DATE` | 日付 | `2026-07-29` |
| 日付 | `DATETIME` | 日付＋時刻 | `2026-07-29 22:40:00` |
| 日付 | `TIMESTAMP` | 日付＋時刻（自動記録可能） | 作成日時 |

## 2. 型変換（型変換関数を使用）-> CAST()関数を使用

#### 1) 文字列 → 数値

```python
%%sql

SELECT CAST('100' AS FLOAT) AS result; 
-- カラムを使用せず、値や計算式だけをSELECT句で取得する場合はFROM句を省略できる
```

##### # 文字列同士の加算による暗黙的な型変換

```python
%%sql

SELECT '100' + '200' AS result;
```

```python
%%sql

SELECT '100' + 50;
```

```python
%%sql

SELECT '100円' + 50;
```

```python
%%sql

SELECT 'xyz' + 50;  -- 文字列は0に変換される
```

#### 2) 数値 → 文字列

```python
%%sql

SELECT CAST(123 AS CHAR);
```

```python
%%sql

SELECT CONCAT(CAST(100 AS CHAR), '円');
```

```python
%%sql

SELECT
    CONCAT(CAST(customer_id AS CHAR), '番') AS 番号,
    first_name
FROM customer
LIMIT 5;
```

```python
%%sql

-- 暗黙的な型変換を利用
SELECT
    CONCAT(customer_id, '番') AS 番号,
    first_name
FROM customer
LIMIT 5;
```

#### 3) 文字列 → 日付

```python
%%sql

SELECT CAST('2026-07-29' AS DATE);
```

```python
%%sql


SELECT CAST('2026-07-29 22:40:00' AS DATETIME);
```

#### 4) 日付 → 文字列

```python
%%sql

SELECT CAST(NOW() AS CHAR);
```

#### 5) 文字列 → 日付

```python
%%sql

SELECT STR_TO_DATE('20260729', '%Y%m%d');
```

```python
%%sql

SELECT STR_TO_DATE('2026-07-29', '%Y-%m-%d');
```

#### 6) 実数 ↔ 整数

```python
%%sql

SELECT CAST(123.987 AS SIGNED);
```

```python
%%sql

SELECT CAST(100 AS DECIMAL(10,2));
```

```python
%%sql

SELECT
    CAST(10 / 3 AS UNSIGNED),
    CAST(10 / 3 AS DECIMAL(10, 2)),
    CAST(10 / 3 AS FLOAT),   -- 単精度浮動小数点数（約7桁）
    CAST(10 / 3 AS DOUBLE);  -- 倍精度浮動小数点数（約15桁）
```
