# 13) 数値・文字列・日付型の単一行関数

MySQLのすべての関数については、[公式ドキュメント](https://dev.mysql.com/doc/refman/8.4/en/functions.html)を参照する。

- 単一行関数（Single Row Function）は、1行のデータを受け取り、1行の結果を返す関数である。<br>例えば、顧客が599人いる場合、関数も599回実行され、599件の結果を返す。

## 1. 文字列の単一行関数

| 関数 | 説明 | 例 |
|---|---|---|
| `UPPER()` | 文字列を`大文字`に変換する | `UPPER(name)` |
| `LOWER()` | 文字列を`小文字`に変換する | `LOWER(name)` |
| `LENGTH()` | 文字列の`バイト数`を取得する | `LENGTH(name)` |
| `CONCAT()` | 複数の文字列を`連結`する | `CONCAT(a, b)` |
| `LEFT()` | 文字列の`左側`から指定した文字数を取得する | `LEFT(name, 2)` |
| `RIGHT()` | 文字列の`右側`から指定した文字数を取得する | `RIGHT(name, 2)` |
| `SUBSTRING()` | 指定した位置から`文字列の一部`を取得する | `SUBSTRING(name, 2, 3)` |
| `TRIM()` | 文字列の`前後の空白`を削除する | `TRIM(name)` |
| `REPLACE()` | 文字列の一部を別の文字列に`置換`する | `REPLACE(email, '.com', '.net')` |

```python
%%sql

SELECT
    first_name,
    UPPER(first_name) AS upper_name
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    first_name,
    LOWER(first_name) AS lower_name
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    first_name,
    LENGTH(first_name) AS length
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    email,
    REPLACE(email, '.org', '.com') AS new_email
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    first_name,
    LEFT(first_name, 3) AS first3
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    first_name,
    RIGHT(first_name, 3) AS last3
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    first_name,
    SUBSTRING(first_name, 2, 3) AS sub_name
FROM customer
LIMIT 5;
```

```python
%%sql

-- 顧客の名前と姓を組み合わせ、名前の長さを一緒に出力
SELECT
    CONCAT(first_name, ' ', last_name) AS full_name,
    LENGTH(first_name) AS name_length
FROM customer
LIMIT 5;
```

## 2. 数値の単一行関数

| 関数 | 説明 |
|---|---|
| `ROUND()` | 四捨五入 |
| `TRUNCATE()` | 切り捨て |
| `CEIL()` | 切り上げ |
| `FLOOR()` | 切り下げ |
| `ABS()` | 絶対値 |
| `MOD()` | 余り |
| `RAND()` | 乱数 |

```python
%%sql

SELECT
    payment_id,
    amount,
    ROUND(amount, 1) AS round_amount
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    TRUNCATE(amount, 1) AS truncate_amount
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    CEIL(amount) AS ceil_amount
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    FLOOR(amount) AS floor_amount
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    ABS(amount - 5) AS absolute_value
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    customer_id,
    MOD(customer_id, 10) AS remainder 
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    customer_id,
    FLOOR(RAND() * 100) + 1 AS coupon_number
FROM customer
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    ROUND(amount * 0.9, 2) AS discount_price
FROM payment
LIMIT 5;
```

```python
%%sql

SELECT
    payment_id,
    amount,
    amount * 0.1 AS discount,
    round(amount-(amount * 0.1), 2) AS discount_price
FROM payment
LIMIT 5;
```
