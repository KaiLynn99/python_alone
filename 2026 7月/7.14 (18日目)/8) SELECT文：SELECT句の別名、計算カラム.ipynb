## 1. 別名

`AS`は、カラムやテーブルに別名（Alias）を付けるためのキーワードである。

#### - カラム名に別名を付ける

```sql
SELECT
    first_name AS 名,
    last_name AS 姓
FROM customer;
```

#### - `AS`は省略可能

```sql
SELECT
    first_name 名
    last_name 姓
FROM customer;
```

### - テーブルに別名を付ける

- カラムの別名：取得結果を分かりやすくするために使用する。
- テーブルの別名：記述を簡潔にし、可読性を高めるために使用する。

```sql
SELECT
    c.customer_id,
    c.first_name
FROM customer AS c;  -- ASは省略可能：FROM customer c
```

## 2. 計算カラム

## 2. 計算カラム

`SELECT句`では、既存のカラムだけでなく、計算した値を新しいカラム（計算カラム、Derived Column）として取得できる。

#### - 顧客番号に1000を加える

````python
%%sql

SELECT
    customer_id,
    customer_id + 1000 AS new_id
FROM customer
LIMIT 5;
````

#### - カラムを関数の引数として使用する

- カラムを関数の引数として使用し、その戻り値を新しいカラムとして取得できる。  

````python
%%sql

SELECT
    first_name,
    last_name,
    CONCAT(first_name, ' ', last_name) AS full_name
FROM customer
LIMIT 5;
````

#### 特定の値を取得結果に追加する

- 特定の値（計算値、関数の戻り値など）を取得結果に追加できる。

````python
%%sql

SELECT
    customer_id,
    first_name, 
    10*10,
    'SQL',
    NOW()
FROM customer 
LIMIT 5;
````

※ `ORDER BY句`は`SELECT句`の後に実行されるため、`SELECT句`で指定した別名やカラムの位置番号を使用できる。

`WHERE句`は`SELECT句`より先に実行されるため、`SELECT句`で指定した別名は使用できない。

````python
%%sql

SELECT
    first_name AS 名,
    last_name AS 姓,
    customer_id % 2 AS 余り
FROM customer
ORDER BY first_name 
LIMIT 10;

-- SELECT句より後に実行されるため、別名または位置番号を使用できる
-- ORDER BY 名;
-- ORDER BY 1;
````
