```python
%%bigquery

WITH score_data AS (
    SELECT '국어' AS subject, 90 AS score UNION ALL
    SELECT '영어', 85 UNION ALL
    SELECT '수학', 100
)
SELECT *
FROM score_data
PIVOT (
    MAX(score)
    FOR subject IN ('국어', '영어', '수학')
);
```
