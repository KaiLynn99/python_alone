## 1. 拡張GROUP BY

`GROUP BY`には、基本的な`GROUP BY`のほかに複数の集計結果を一度に取得するための拡張機能がある。

```text
GROUP BY
│
├─ 基本 GROUP BY
│
└─ 拡張 GROUP BY（Advanced / Extended GROUP BY）
    ├─ ROLLUP
    ├─ CUBE
    ├─ GROUPING SETS
    └─ GROUPING()
```

| 機能 | 説明 |
|---|---|
| `ROLLUP` | 小計・総計を自動生成 |
| `CUBE` | すべての組み合わせで集計 |
| `GROUPING SETS` | 必要な集計パターンだけを指定 |
| `GROUPING()` | 集計による`NULL`と実データの`NULL`を区別 |

## 3. 拡張 GROUP BY
