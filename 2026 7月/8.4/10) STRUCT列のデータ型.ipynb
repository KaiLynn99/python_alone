## 1. STRUCTデータ型

`STRUCT`は、複数の項目を一つの列にまとめて保存するデータ型である。  
Pythonの`dict`型に似ている。

```python
%%bigquery

WITH tmp AS (
    SELECT
        STRUCT(
            '千尋' AS name,
            20 AS age,
            '東京' AS city
        ) AS person,
        'JAPAN' AS country

    UNION ALL

    SELECT
        STRUCT(
            'ハウル' AS name,
            25 AS age,
            '京都' AS city
        ) AS person,
        'JAPAN' AS country
)

SELECT
    person,
    country
FROM tmp;

-- `person`列の中に`name`・`age`・`city`がまとめて保存される。
-- `country`は`person`とは別の一般列である。
```

#### 1）ドット演算子でSTRUCT内の値を個別に取得する

```python
%%bigquery

WITH tmp AS (
    SELECT
        STRUCT(
            '千尋' AS name,
            20 AS age,
            '東京' AS city
        ) AS person,
        'JAPAN' AS country

    UNION ALL

    SELECT
        STRUCT(
            'ハウル' AS name,
            25 AS age,
            '京都' AS city
        ) AS person,
        'JAPAN' AS country
)

SELECT
    person.name,
    person.city,
    country
FROM tmp;
```
