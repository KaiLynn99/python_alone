## 1. ORDER BY句

`ORDER BY`句は、取得したデータを指定したカラムの値で並べ替えるために使用する。

```sql
SELECT カラム名
FROM テーブル名
WHERE 条件
ORDER BY カラム名 ASC または DESC;

````python
%%sql

SELECT *
FROM customer
ORDER BY customer_id ASC
LIMIT 3;
````

````python
%%sql

SELECT *
FROM customer
ORDER BY customer_id DESC
LIMIT 3;
````

````python
%%sql

-- -- 文字列の値でも並べ替えることができる。
SELECT *
FROM customer
ORDER BY first_name
LIMIT 3;
````

````python
%%sql create_date_result << 

-- 日付の値でも並べ替えることができる。
SELECT *
FROM customer
ORDER BY create_date
````

````python
save_result_csv(
    create_date_result,
    "7) ORDER/create_date_result.csv"
)
````

````python
%%sql

SELECT *
FROM customer
ORDER BY last_name, first_name
LIMIT 5;
-- last_nameを昇順で並べ替え、
-- last_nameが同じ場合はfirst_nameを昇順で並べ替える。
````

````python
%%sql

-- 最近登録された顧客順
SELECT *
FROM customer
ORDER BY create_date DESC
LIMIT 5;
````
