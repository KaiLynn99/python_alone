データベースは、**最小限の意味を持つテーブル**で構成され、各テーブルの列（カラム）を通じて関連付けられる。

データを保存する際は、データを意味ごとにテーブルへ分割し、以下のキーを使ってテーブル間に関係を持たせる。

- `PRIMARY KEY（主キー）`：各行を一意に識別するキー
- `FOREIGN KEY（外部キー）`：他のテーブルの主キーを参照し、テーブル同士を関連付けるキー
- 関連するデータを取得する場合は`JOIN`を使用する。

#### ※  **リレーション（Relation）**：2つのテーブル間に存在する業務上の関連性



## ※ リレーションとテーブルの違い

- `リレーション`：データ間の関係を表す`論理的`な概念。
- `テーブル`：リレーションを実際に表現する`具体的`な形式。

※ 1つのリレーションを、必ずしも1つのテーブルだけで表現する必要はない。  
重複や異常現象を防ぐために複数のテーブルに分割し、必要なときに`JOIN`で再構成することができる。

#### # テーブル設計のポイント

- データの重複が少ない設計が望ましい。
- ただし、外部キーなどの制約が増えると、データ収集や処理が複雑になる場合もある。
- 目的に応じて、管理のしやすさ・性能・コストとのバランスを考える必要がある。

#### 例)：履修データ

「学生・履修・科目」の情報は、

- 1つの履修テーブルとして設計する方法
- **学生**・**履修**・**科目**の3つのテーブルに分割する方法

のどちらでも表現できる。

ただし、1つのテーブルにまとめるとデータの重複や異常現象が発生しやすいため、必要に応じて正規化する。
