## ※ 整合性

データの`正確性・有効性`を保ち、不正な変更や破壊を防ぐことである。

#### 1) ドメイン整合性（Domain Integrity）

- 各属性には`有効な値のみ`を入力できるようにする
- 例：
  - 郵便番号 → 決められた桁数・形式のみ
  - 商品数量 → 負の値を入力できないように制限

#### 2) エンティティ整合性（Entity Integrity）

- 各行を`一意に識別`できるようにする
- `主キーはNULL不可`

#### 3) 参照整合性（Referential Integrity）

- 他のテーブルを参照する場合、存在する値を正しく参照しなければならない
- `外部キー`は`NULL`、または参照先の`主キーに存在する値`を持つ

> 例：学生テーブルが学科テーブルを参照する場合、存在しない学科を参照すると参照整合性が崩れる。

#### 4) 正規化整合性（Normalization Integrity）

- `原子性`や各種の`従属性`など、各正規形の条件に基づいてリレーションを分解する
- `異常（Anomaly）`を取り除いた望ましいリレーションにする

#### 例) 部署情報が社員ごとに重複している場合

| 社員番号 | 社員名 | 部署番号 | 部署名 |
|---|---|---|---|
| 1001 | 千尋 | 10 | 開発部 |
| 1002 | ハウル | 10 | 開発部 |

`部署番号 → 部署名`という`関数従属性`があるため、テーブルを分割する。

- 社員：`社員番号・社員名・部署番号`
- 部署：`部署番号・部署名`

→ 重複を減らし、`更新・挿入・削除異常`を防ぐ。

### 整合性制約（Integrity Constraints）

| 整合性制約 | 役割 |
|---|---|
| `NOT NULL` | そのカラムに`NULL`を許可しない |
| `UNIQUE` | そのカラムの値を`一意`にし、重複を許可しない |
| `PRIMARY KEY` | 値が必ず存在し、一意でなければならない（`NOT NULL + UNIQUE `） |
| `FOREIGN KEY` | そのカラムの値が参照先のカラムの値を`参照`する |
| `CHECK` | そのカラムに保存できる値が`指定した範囲・条件`を満たすか`検査`する |
