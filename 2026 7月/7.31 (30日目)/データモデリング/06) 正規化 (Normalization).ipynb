## 正規化とは

`異常（anomaly）`の原因となるデータの重複を最小限にするため、一定の制約条件に従って`リレーション（テーブル）`を分割する作業である。

## 1. 正規化の基本原理

- 一つのリレーション内における、属性間の関係を効率的に表現する。
- 不要なデータの重複を防止・除去する。
- 新しい属性をデータベースに追加する際、既存の属性との関係の修正を最小限にする。
- データ保存に必要な記憶領域の無駄を減らし、データベースの情報品質を向上させる。

正規化の基本的な考え方は、`基本キー`と論理的な従属関係を持つ属性をまとめて、一つのリレーションを作成することである。

データベース設計者は、各リレーションに存在する`関数従属性`を確認し、リレーション全体が特定の正規形を満たすようにテーブルを分割する。

正規化によって作成されたリレーションスキーマは、データを登録・更新・削除する際に発生する情報の不整合を防止できる。

## ※ 2. リレーショナルデータベース設計の指針

| 指針 | 説明 | 例 |
|---|---|---|
| **指針1：分かりやすく明確なスキーマを作成する** | 一つのテーブルでは、一つの`主題（エンティティ）`だけを表現するように設計する。 | **悪い例：** 社員テーブルに部署名や部署所在地まで保存する。<br><br>**改善例：** `社員（Employee）`テーブルと`部署（Department）`テーブルに分離する。 |
| **指針2：NULL値をできるだけ避ける** | `NULL`が多くなると、データの管理や検索が難しくなる。必要に応じてテーブルを分離し、NULL値を減らす。 | 学生テーブルに`卒業日`列がある場合、在学生の卒業日はすべてNULLになる。<br><br>そのため、学生情報と卒業情報を別々のテーブルで管理する。 |
| **指針3：偽のタプルが発生しないようにする** | テーブルを誤って分割すると、`JOIN`した際に実際には存在しないデータの組み合わせが生成される場合がある。 | 学生とサークルの所属情報を誤って分割すると、JOIN時に実際には所属していない学生とサークルが結び付く可能性がある。 |
| **指針4：スキーマを整理する** | `関数従属性`を利用して重複データを除去し、`第1正規形（1NF）`、`第2正規形（2NF）`、`第3正規形（3NF）`などを満たすように設計する。 | 社員テーブルに部署名と部署所在地が繰り返し保存されている場合、部署情報を別の部署テーブルに分離して重複を除去する。 |

## 2. 正規化の目的

正規化の主な目的は、次のとおりである。

1. データの重複を最小限にする。
2. データの挿入・更新・削除時に発生する異常を防止する。
3. データの一貫性と整合性を維持する。
4. 属性間の依存関係を明確にする。
5. データベースの修正や拡張を容易にする。

## 3. 第1正規形〜第5正規形

正規化では、`第1正規形（1NF）`から`第5正規形（5NF）`までが代表的である。

このほかに、`BCNF（Boyce-Codd Normal Form：ボイス・コッド正規形）`も扱われる。

| 正規形 | 主な目的 | 実務での使用頻度 |
|---|---|:---:|
| **1NF** | 各列に`原子的な値`だけを保存し、繰り返し列を除去する | ★★★★★ |
| **2NF** | `部分関数従属`を除去する | ★★★★★ |
| **3NF** | `推移的関数従属`を除去する | ★★★★★ |
| **BCNF** | 3NFより厳しく、すべての決定項が候補キーになるようにする | ★★☆☆☆ |
| **4NF** | `多値従属（Multivalued Dependency）`を除去する | ★☆☆☆☆ |
| **5NF** | `結合従属（Join Dependency）`を除去する | ★☆☆☆☆ |

- 実務では、多くの場合`第3正規形（3NF）`までで十分である。  
- BCNF・4NF・5NFは、特殊なデータモデルで使用されることが多く、一般的なシステムでの使用頻度は低い。  

## 4. データベースを正規化すると

データの一貫性を維持し、`重複を最小限`にして`データの安定性を確保`するために、一つのテーブルを二つ以上に分割する。

その結果、保存領域を効率化し、データの整合性を高めることができる。

## 5. 正規化とNULL除去

#### # 正規化

- データの重複を減らし、`挿入異常・更新異常・削除異常`を防ぐためにテーブルを分割する設計技法である。

#### # NULL除去

- 一つのテーブルに多くの`NULL`が発生しないように、列を分離したりテーブル構造を変更したりする設計方法である。

- 正規化の過程でNULLが減る場合もあるが、NULLを減らすための列分割が必ずしも正規化になるわけではない。

> **正規化の目的：** 重複の削減、データの整合性向上  
> **NULL除去の目的：** 管理性、保存効率、データ品質の向上
