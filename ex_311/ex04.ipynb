```python
import pandas as pd
```

```python
t = pd.Series(
    [-2.4, 0.4, 5.7, 12.5, 18.2, 23.0,
     25.8, 26.1, 21.2, 14.8, 7.2, 0.1],
)

t.index = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
```

```python
t
```

```python
t.iloc[0] = -3.4
t
```

```python
# 2개의 인덱스 값 변경

t.iloc[6:8] = [30.0, 35.0]
t
```

```python
# 새로운 인덱스 추가

t.loc["AVG"] = t.mean()
t
```

```python
# 인덱스 삭제

t2 = t.drop("AVG")
t2
```
