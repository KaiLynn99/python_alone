```python
import pandas as pd

score = pd.Series(
    [78, 85, 82, 90, 88, 92],
    index=["1학년중간", "1학년기말",
           "2학년중간", "2학년기말",
           "3학년중간", "3학년기말"]
)

print(score)
```

문제 1. Series 기본 정보

다음 정보를 출력하라.

데이터 개수
shape
데이터 타입
인덱스
값

```python
print("데이터 개수 :", score.size)
print("shape :", score.shape)
print("데이터 타입 :", score.dtype)
print("인덱스 :", score.index)
print("값 :", score.values)
```

문제 2. 학년별 성적 조회

loc를 이용하여 다음을 출력하라.

1학년 중간고사 성적
2학년 기말고사 성적
3학년 중간고사 성적

```python
print("1학년 중간고사 성적 :", score.loc["1학년중간"])
print("2학년 기말고사 성적 :", score.loc["2학년기말"])
print("3학년 중간고사 성적 :", score.loc["3학년중간"])
```

문제 3. iloc를 이용한 조회

iloc를 이용하여 다음을 출력하라.

첫 번째 시험 성적
마지막 시험 성적
1학년 두 시험 성적
3학년 두 시험 성적

```python
print("첫 번째 시험 성적 :", score.iloc[0])
print("마지막 시험 성적 :", score.iloc[5])
print("1학년 두 시험 성적 :", score.iloc[0:2])
print("3학년 두 시험 성적 :", score.iloc[4:6])
```

문제 4. 성적 슬라이싱

iloc를 이용하여 다음을 출력하라.

처음 3개의 시험 성적
마지막 3개의 시험 성적
짝수 번째 위치의 성적
역순으로 성적 출력

```python
print("처음 3개의 시험 성적 :", score.iloc[:3])
print()
print("마지막 3개의 시험 성적: ", score.iloc[:2:-1])
print()
print("짝수 번째 위치의 성적: ", score.iloc[::2])
print()
print("역순으로 성적 출력: ", score.iloc[::-1])
```
