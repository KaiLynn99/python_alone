```python
import pandas as pd
```

```python
# 1. 딕셔너리로 생성

# {레이블 : [], 레이블 : []}
df1 = pd.DataFrame({
    "이름" :["민수", "지영"],
    "점수" :[90, 85]
    })
# print() 터미널 출력
# display() 노트북 셀에 최적화 출력
df1
```

```python
# 2. 리스트이 리스트로 생성 

# index값 행레이블(레이블 인덱스), columns값 열레이블(레이블 인덱스)
# 생략하면 위치값으로 기본값이 설정된다.

df2 = pd.DataFrame([
    ["민수", 90], # 1행
    ["지영", 100], # 2행
], index=["첫번째", "두번째"], columns=["이름", " 점수"])
df2
```

```python
# 3. 리스트의 딕셔너리로 생성

df3 = pd.DataFrame([
     {"이름":"민수", "점수":90}, # 1행
     {"이름":"지영", "점수":100}, # 2행
])
df3
```

```python
# 4. Numpy ndarray로 생성

import numpy as np

# numpy의 2차원 배열 - 모든 데이터 타입이 동일해야 함

df4 = pd.DataFrame(np.array([
    [1,2],
    [3,4],
]), columns=["x", "y"])

df4
```

```python
# 이름, 나이, 성별 데이터로 DataFrame 생성
df = pd.DataFrame({
    '이름': ['철수', '영희', '민수', '지영', '준호'],
    '나이': [20, 21, 20, 22, 21],
    '성별': ['남', '여', '남', '여', '남']
})

df
```
