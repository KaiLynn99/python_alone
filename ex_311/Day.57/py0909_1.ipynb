```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88]
}, index=['A', 'B', 'C', 'D', 'E'])

print(df)
```

```python
df.loc["B"] # df.loc["B", :]
```

```python
df.loc["B", "이름"]
```

```python
df.iloc[0,2] # index레이블이 숫자로 되어있다면
```

```python
# [열 이름], [열 이름의 리스트] -> 열 선택
# [식] -> 행 선택
# [:]
# [조건식]
```

```python
# t = df["점수"]
# df[ t > 90]

df[df["점수"] > 90] # -> 열 값으로 바꾼 후 비교
df["점수" > 90] # -> 오류 (식이므로 문자와 숫자는 계산 안됨)
```
