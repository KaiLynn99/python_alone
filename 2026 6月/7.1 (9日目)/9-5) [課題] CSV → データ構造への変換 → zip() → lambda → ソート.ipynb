## 問題 2

#### 生徒の

#### 名前リスト 

#### 国語リスト 

#### 英語リスト 

#### をそれぞれ保存した後、出力しなさい。

#### 出力例）

```python
print(names) # ['キキ', 'アシタカ', ...]
print(langs)
print(engs)         
```

````python
names = []
langs = []
engs = []

import csv 

with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)

    next(reader)
   
    for line in reader: 
        names.append(line[0])
        langs.append(line[1])
        engs.append(line[2])

print(names)
print(langs)
print(engs)
````

## 問題 3

#### zip() を使用して、

#### 生徒の名前と国語の点数を [(), (), ..] の形式で保存し、出力する

#### 例）

```python
[
('キキ',90),
('アシタカ',88)
]
```

````python
import csv

names = []
langs = [] 

with open("students.csv", "r", encoding="utf-8") as f:

    reader = csv.reader(f)
    next(reader)

    for line in reader:
        names.append(line[0])
        langs.append(int(line[1]))

result = zip(names, langs)

print(list(result))
````

## 問題 5

#### dict(zip()) を使用して、

#### 次のような辞書（2つ：name_kor_dict、name_eng_dict）を保存し、出力する

#### 例）

```python
{'キキ': 90, 
'アシタカ': 88, 
... 
}
\n
{'キキ': 85, 
'アシタカ': 92, 
...
}
```

````python
import csv

names = []
langs = []
engs = []

with open("students.csv", "r", encoding="utf-8") as f:
    
    reader = csv.reader(f)
    next(reader)

    for line in reader:
        names.append(line[0])
        langs.append(int(line[1]))
        engs.append(int(line[2]))
    
name_lang_dict = zip(names, langs)
name_eng_dict = zip(names, engs)

print(dict(name_lang_dict))
print(dict(name_eng_dict))
````

## 問題 6

#### 国語の点数の高い順に並べ替えてください。
#### (stored()、ラムダ式を使用)

例）
```python
[
('キキ',100),
('アシタカ',95),
...
]
```

````python
import csv

names = []
langs = []

with open("students.csv", "r", encoding="utf-8") as f:
    
    reader = csv.reader(f)
    next(reader)

    for line in reader:
        names.append(line[0])
        langs.append(int(line[1]))

result = list(zip(names, langs))

answer = sorted(result, key=lambda x: x[1], reverse=True)
print(answer)
````

## 問題 10

#### 問題8の結果から、平均点が90点以上の生徒のみを出力しなさい。
#### (filter()、ラムダ式を使用)

#### 例）
```python
キキ 97.5
アシタカ 99.0
サン 93.5
```

````python
import csv

names = []
lang_eng_aver = []

with open("students.csv", "r", encoding="utf-8") as f:
    
    reader = csv.reader(f)
    next(reader)

    for line in reader:
        names.append(line[0])
        lang_eng_aver.append((int(line[1])+int(line[2])) /2 )

result = zip(names, lang_eng_aver)

answer = list(filter(lambda x: x[1] >= 90, result))
print(answer)
````
