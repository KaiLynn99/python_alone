## 2. sklearnのKBinsDiscretizerクラスを使用した離散化の実習

```python
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer

# データ
df = pd.DataFrame({
    "年齢": [15, 23, 35, 47, 62]
})

display(df)

# 離散化
kb = KBinsDiscretizer(
    n_bins=3,             # 3つの区間
    encode="ordinal",     # 区間番号を数値で表現
    strategy="uniform"    # 等間隔で区間を分割
)

# strategyの指定方法は3種類
# uniform   → 区間の幅を均等にする
# quantile  → 各区間のデータ数がほぼ同じになるように分割
# kmeans    → K-Means方式で区間を決定

df["年齢_区間"] = kb.fit_transform(df[["年齢"]])

display(df)
```

※ `uniform`では、年齢の範囲を<strong>同じ幅で3つの区間</strong>に分ける。  
`0.0 / 1.0 / 2.0` は年齢そのものではなく、<strong>所属する区間番号</strong>を表す。

- 15, 23 → 区間 0
- 35 → 区間 1
- 47, 62 → 区間 2
