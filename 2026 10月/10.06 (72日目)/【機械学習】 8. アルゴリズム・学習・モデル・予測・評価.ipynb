## 2. 学習とは

**アルゴリズムに与えられたデータを使って、データ内の規則を見つける過程**。

<div style="display:inline-block; min-width:620px; font-family:Consolas, 'Yu Gothic', monospace; line-height:28px; margin:4px 0 14px 0; color:#222;"><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;"><strong>例）勉強時間と試験点数</strong></div><div style="background:#f4f5f6; padding:0 14px;">2時間 → <span style="color:#ff5a1f;"><strong>60点</strong></span></div><div style="background:#e7e9eb; padding:0 14px;">4時間 → <span style="color:#ff5a1f;"><strong>70点</strong></span></div><div style="background:#f4f5f6; padding:0 14px;">6時間 → <span style="color:#ff5a1f;"><strong>85点</strong></span></div><div style="background:#e7e9eb; padding:0 14px;">8時間 → <span style="color:#ff5a1f;"><strong>95点</strong></span></div><div style="background:#f4f5f6; padding:0 14px;">→ 「勉強時間が増えると試験点数も上がる」という規則を見つける過程が<strong>学習</strong>である。</div></div>
