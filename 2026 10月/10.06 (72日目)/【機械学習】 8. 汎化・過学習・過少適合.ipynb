## 1. 汎化

- 学習したモデルが、**初めて見る新しいデータに対しても正しく予測できる能力**である。
- 新しいデータに対してもうまく予測できる場合、**汎化性能が高いモデル**という。
- **機械学習の究極的な目標 = 汎化**

## 2. 過学習（Overfitting）

**訓練データに過度に適合している状態。**

<div style="background:#ffffff; padding:4px 18px; white-space:pre; color:#111 !important;">    ↓</div>

<div style="background:#f2f3f4; padding:4px 18px; white-space:pre; color:#111 !important;">学習しすぎる</div>

<div style="background:#ffffff; padding:4px 18px; white-space:pre; color:#111 !important;">    ↓</div>

<div style="background:#f2f3f4; padding:4px 18px; white-space:pre; color:#111 !important;">訓練データにはよく適合できる</div>

<div style="background:#f2f3f4; padding:4px 18px; white-space:pre; color:#111 !important;">    ↓</div>

##### ■ **発生原因**

- モデルが**複雑すぎる**
- 学習データが不足している
- 不要な特徴量が多い
- 学習を長く行いすぎる  
- 正則化が不足している

<br>

##### ■ **例)** 

訓練精度 = 99% / テスト精度 = 75% <br>
→ **過学習の可能性が高い。**

<div style="background:#f2f3f4; padding:4px 18px; white-space:pre; color:#111 !important;">訓練データにも十分に適合できない</div>

##### ■ **発生原因**

- モデルが**単純すぎる**
- 学習が十分ではない
- 重要な特徴量が不足している
- 正則化が強すぎる

<br>

##### ■ **例**

訓練精度 = 60% / テスト精度 = 55%  
→ **過少適合の可能性が高い。**
