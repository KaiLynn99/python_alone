## 1. 主成分分析（PCA）

主成分分析（PCA）は、**データの特徴をできるだけ維持しながら、変数の数を減らす次元削減の方法**である。

例えば、「身長・体重・年齢」の3つの変数を、より少ない主成分にまとめることができる。

<div style="display:inline-block; min-width:820px; font-family:Consolas, 'Yu Gothic', monospace; line-height:28px; margin:4px 0 14px 0; color:#222;"><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;"><strong>主成分分析</strong>は、既存の変数に<strong>重みを付けて新しい変数を作る方法</strong>である。</div><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;">例として、<span style="color:#ff5a1f;"><strong>3個の変数</strong></span>を<span style="color:#ff5a1f;"><strong>2個の主成分</strong></span>に変換すると仮定する。</div><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;"><strong>［主成分1］新しい</strong><span style="color:#ff5a1f;"><strong> X1</strong></span></div><div style="background:#f4f5f6; padding:0 14px;">身長 × <span style="color:#ff5a1f;">0.7</span> ＋ 体重 × <span style="color:#ff5a1f;">0.2</span> ＋ 年齢 × <span style="color:#ff5a1f;">0.1</span></div><div style="background:#e7e9eb; padding:0 14px;">&nbsp;</div><div style="background:#f4f5f6; padding:0 14px;"><strong>［主成分2］新しい</strong><span style="color:#ff5a1f;"><strong> X2</strong></span></div><div style="background:#e7e9eb; padding:0 14px;">身長 × <span style="color:#ff5a1f;">0.2</span> ＋ 体重 × <span style="color:#ff5a1f;">0.5</span> ＋ 年齢 × <span style="color:#ff5a1f;"> (-0.8)</span></div><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;">※ 実際のPCAでは、データの分散を分析して<strong>重みを自動的に計算</strong>する。上記の重みは理解のための例。</div><div style="background:#f4f5f6; padding:0 14px;">→ 元の<span style="color:#ff5a1f;"><strong>3個の変数</strong></span>を組み合わせ、新しい<span style="color:#ff5a1f;"><strong>2個の変数</strong></span>へ削減する。</div></div>

<br>

### 実習) 身長・体重・年齢の3変数を、PCAによって2つの主成分に変換する

<br>

#### ① サンプルデータの作成

```python
import pandas as pd
from sklearn.decomposition import PCA

df = pd.DataFrame({
    "身長": [160, 165, 172, 168, 180],
    "体重": [55, 63, 60, 75, 68],
    "年齢": [20, 35, 25, 45, 30]
})

display(df)
```

<br>

#### ② PCAで3変数を2変数へ削減

```python
pca = PCA(n_components=2)   # 元の3つの変数を、2つの主成分に変換する

X_pca = pca.fit_transform(df)
```

<br>

#### ③ PCA変換後の結果を確認

```python
df_pca = pd.DataFrame(
    X_pca,
    columns=["主成分1", "主成分2"]
)

display(df_pca)
```

<br>

#### ④ 各主成分の説明率を確認

```python
display(pca.explained_variance_ratio_)
```

<div style="display:inline-block; min-width:760px; font-family:Consolas, 'Yu Gothic', monospace; line-height:28px; margin:4px 0 14px 0; color:#222;"><div style="background:#f4f5f6; padding:0 14px;">&nbsp;</div><div style="background:#e7e9eb; padding:0 14px;">説明率：［<span style="color:#ff5a1f;"><strong>0.726</strong></span>, <span style="color:#ff5a1f;"><strong>0.264</strong></span>］</div><div style="background:#f4f5f6; padding:0 14px;">主成分1 → データの特徴を約 <span style="color:#ff5a1f;"><strong>73%</strong></span> 説明</div><div style="background:#e7e9eb; padding:0 14px;">主成分2 → データの特徴を約 <span style="color:#ff5a1f;"><strong>26%</strong></span> 説明</div><div style="background:#f4f5f6; padding:0 14px;">→ 2つの主成分だけで、元のデータの特徴を約 <span style="color:#ff5a1f;"><strong>99%</strong></span> 説明できる。</div></div>
