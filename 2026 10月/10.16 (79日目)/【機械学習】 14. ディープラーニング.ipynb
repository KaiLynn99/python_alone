## 2. 機械学習：アルゴリズムを選択

```python
from sklearn.neighbors import KNeighborsClassifier

# KNNモデルの作成
model = KNeighborsClassifier(n_neighbors=5)

# モデルの学習
model.fit(X_train, y_train)
```

- `KNeighborsClassifier()`でKNNモデルを作成し、`fit()`で学習する。

- Random Forestやロジスティック回帰も、それぞれ専用のモデルクラスを使用する。

<br>

## 3. ディープラーニング：ニューラルネットワークを設計

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

# ニューラルネットワークモデルの作成
model = Sequential()

# 隠れ層を追加
model.add(Dense(100, activation="relu"))

# 出力層を追加
model.add(Dense(3, activation="softmax"))
```

- `Sequential`：層を順番に接続
- `Dense`：前の層とすべてのニューロンを接続する全結合層
- `100`：その層のニューロン数
- `relu`：隠れ層でよく使用する活性化関数
- `softmax`：多クラス分類で使用する出力層の活性化関数

→ このように、ディープラーニングではニューラルネットワークの構造を設計する過程が重要である。

<br>

## 4. ディープラーニングの学習設定

```python
# 学習設定
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# モデルの学習
model.fit(
    X_train,
    y_train,
    epochs=100
)
```

- **損失関数（Loss Function）**：予測と正解の差を計算
- **最適化アルゴリズム（Optimizer）**：損失を小さくするように重みを更新
- **Epoch**：全学習データを1回学習する単位

(`model.fit()` の内部では、**逆伝播（Backpropagation）**によって誤差から各重みの修正量を計算する)

#### ※ 機械学習は今でも使われる？

- ディープラーニングが強力でも、すべての問題で最適とは限らない。

- 表形式データでは、ロジスティック回帰やRandom Forestなどの機械学習モデルがよく使用される。  

- 画像・音声・テキストなど複雑なデータでは、ディープラーニングが有効な場合が多い。

→ **データや目的に応じて、機械学習とディープラーニングを使い分ける。**
