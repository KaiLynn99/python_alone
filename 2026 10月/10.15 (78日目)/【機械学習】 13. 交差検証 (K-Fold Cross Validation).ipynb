## 1. `交差検証`とは、

- データを複数に分けて**学習と評価を繰り返し、モデルの性能を安定して評価する方法である**。

- モデルの性能を自動的に向上させる方法ではなく、データ分割による偶然の影響を減らし、**より信頼性の高い性能評価**を行うために使用する。

- 交差検証の平均スコアは、**ホールドアウト法より安定して汎化性能を推定するための値**。<br> ※ ホールドアウト法：データを1回だけ学習用・評価用に分割する方法。

<br>

## 2. 交差検証を行う理由

<div style="color:#ff0000; line-height:1.9;">

<strong>1) 1回だけのTrain/Test分割</strong>では、テストデータの特徴によって評価結果が偶然偏る可能性があり、<strong>モデルの本当の汎化性能を十分に判断できない。</strong><br>

<strong>2) 交差検証</strong>では、全データを<strong>K個のFold</strong>に分け、検証用Foldを入れ替えながら<strong>学習と評価を繰り返す</strong>（模擬試験のようなイメージ）。<br>

<strong>3) 複数回の評価結果の平均</strong>を確認することで、モデルが<strong>新しいデータでも安定した性能を発揮できるか</strong>を客観的に評価する。

</div>

## 3. K-Fold交差検証

全データを**K個のFold**に分け、各Foldを1回ずつ検証データとして使用する。

<br>

#### 1) K-Fold交差検証の仕組み

- データ数が少ない場合、固定した検証データだけで評価すると、学習に使えるデータが減り、評価結果も偏る可能性がある。

- K-Fold交差検証では、すべてのデータを順番に**学習用・検証用**として利用するため、データを有効に使いながら、より安定した評価ができる。

- 例えると、**5回の異なる模擬試験の平均点で実力を評価する**ようなもの (1回だけの評価による偶然や外れ値の影響を小さくできる)

#### 2) K-Fold交差検証の流れ

- データを同じ大きさの **K個のFold** に分割
- 1つのFoldを検証用、残りの **K-1個** を学習用として使用
- 検証用のFoldを順番に入れ替えながら、合計 **K回** 学習・評価
- K個の評価スコアを取得
- 最後にスコアの**平均**をモデルの代表的な評価値として使用

### 3) 回帰・分類に応じた分割方法の選択

- **回帰モデル（Regression）**：基本的に **KFold** を使用
- **分類モデル（Classification）**：**StratifiedKFold** を使用

分類では、ターゲットクラスの割合が特定のFoldに偏ると、評価結果も偏る可能性がある。

**StratifiedKFold** は、各Foldでターゲットクラスの割合ができるだけ同じになるように分割する。
