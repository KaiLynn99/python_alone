## 2. テキストの数値化

テキストはそのまま機械学習モデルに入力できないため、**数値表現に変換する必要がある。**

- `CountVectorizer`
  - 単語の**出現回数**を数値化する。

- `TfidfVectorizer`
  - 文書内で**重要な単語ほど高い重み**を与える。

- `Embedding`
  - 単語や文章を**意味や文脈を含むベクトル**として表現する。
